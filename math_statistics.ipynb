# Descriptive Statistics vs Inferential Statistics

Statistics is divided into two main types:

1. **Descriptive Statistics**
2. **Inferential Statistics**

---

## 1. Descriptive Statistics

Descriptive statistics are used to **summarize and describe the data**.

It tells us **what the data looks like**, but it does **not make predictions**.

### Things studied

- Mean (Average)
- Minimum value
- Maximum value
- Variance
- Standard Deviation

### Purpose

To summarize and understand the data.

### Example

Suppose a class has marks:

`70, 75, 80, 85, 90`

Descriptive statistics can tell us:

- Average = 80
- Minimum = 70
- Maximum = 90

It only describes the given data.

---

## 2. Inferential Statistics

Inferential statistics are used to **make predictions or draw conclusions about a population using a sample**.

Instead of studying everyone, we study a small sample and make decisions about the whole population.

### Things studied

- Hypothesis Testing
- Confidence Interval
- Regression

### Purpose

To make predictions or conclusions about a population.

### Example

A college has **10,000 students**.

Instead of asking all students, we survey **200 students**.

If 180 out of 200 students like online classes, we may conclude that **most students in the college like online classes**.

This is called inferential statistics because we used a sample to make a conclusion about the entire population.

---

## Difference Between Descriptive and Inferential Statistics

| Descriptive Statistics | Inferential Statistics |
|------------------------|-------------------------|
| Describes the data | Makes predictions or conclusions |
| Uses sample or population | Uses sample only |
| Summarizes data | Predicts about the population |
| Example: Mean, Min, Max, Variance | Example: Hypothesis Testing, Confidence Interval, Regression |

---

## Real-Life Example

Imagine a school has **1000 students**.

- If we calculate the **average marks of all 1000 students**, it is **Descriptive Statistics**.
- If we check the marks of **100 students** and predict the performance of all 1000 students, it is **Inferential Statistics**.

---

## Key Points

- **Descriptive Statistics = Describe the data.**
- **Inferential Statistics = Predict or conclude about a population using a sample.**
- Descriptive answers **"What does the data look like?"**
- Inferential answers **"What can we predict from the data?"**

# Mean, Median and Mode

Mean, Median, and Mode are three common ways to find the **center** of a dataset.

---

# 1. Mean (Average)

The **Mean** is the **average** of all values.

### Formula

Mean = (Sum of all values) / (Number of values)

### Example

Marks = 70, 80, 90

Mean = (70 + 80 + 90) / 3

Mean = 240 / 3 = **80**

### When to use?

Use the mean when there are **no extreme (very high or very low) values** in the data.

---

# 2. Median

The **Median** is the **middle value** after arranging the data in ascending order.

### Example (Odd number of values)

Data = 10, 20, 30, 40, 50

Median = **30**

(The middle value)

### Example (Even number of values)

Data = 10, 20, 30, 40

There are two middle values: 20 and 30

Median = (20 + 30) / 2 = **25**

### When to use?

Use the median when the data contains **outliers (extreme values)** because it is not affected much by them.

---

# 3. Mode

The **Mode** is the value that appears **most frequently** in the dataset.

### Example

Data = 2, 3, 3, 4, 5

Mode = **3**

because it occurs **2 times**, which is more than any other number.

### When to use?

Mode is useful for **categorical data** or to find the **most common value**.

Example:
- Favorite color of students
- Most purchased product

---

# Comparison

| Measure | Meaning | Example |
|---------|---------|---------|
| Mean | Average of all values | (10 + 20 + 30) / 3 = 20 |
| Median | Middle value after sorting | 10, 20, 30, 40, 50 → 30 |
| Mode | Most frequently occurring value | 2, 3, 3, 4 → 3 |

---

# Real-Life Example

Suppose the ages of five friends are:

18, 20, 20, 22, 25

- **Mean** = (18 + 20 + 20 + 22 + 25) / 5 = **21**
- **Median** = **20**
- **Mode** = **20**

---

# Key Points

- **Mean = Average**
- **Median = Middle value after sorting**
- **If there are an even number of values, Median = Average of the two middle values**
- **Mode = Most frequently occurring value**

# Percentile

A **Percentile** tells us the position of a value in a dataset.

It shows **what percentage of the data is less than or equal to that value**.

Before finding percentiles, the data must be arranged in **ascending order**.

---

## Example

Scores:

38, 41, 45, 50, 54, 65, 66, 69, 78

---

## 25th Percentile (Q1)

The **25th percentile** is the value below which **25% of the data** falls.

In this example:

**25th Percentile = 45**

This means **25% of the scores are less than or equal to 45**.

---

## 50th Percentile (Median or Q2)

The **50th percentile** is also called the **Median**.

It divides the dataset into **two equal halves**.

In this example:

**50th Percentile = 54**

This means **50% of the scores are less than or equal to 54**.

---

## 75th Percentile (Q3)

The **75th percentile** is the value below which **75% of the data** falls.

In this example:

**75th Percentile = 66**

This means **75% of the scores are less than or equal to 66**.

---

## Real-Life Example

Suppose you scored in the **90th percentile** in an exam.

This means:

- You scored **better than 90% of the students**.
- Only **10% of the students scored higher than you**.

---

## Quartiles

- **25th Percentile = First Quartile (Q1)**
- **50th Percentile = Second Quartile (Q2) = Median**
- **75th Percentile = Third Quartile (Q3)**

---

## Key Points

- Percentile shows the **position** of a value in a dataset.
- Data should be **sorted** before finding percentiles.
- **25th Percentile (Q1)** → 25% of data is below or equal to this value.
- **50th Percentile (Q2)** → Median (middle value).
- **75th Percentile (Q3)** → 75% of data is below or equal to this value.
```

# Measures of Dispersion: Range and IQR (Interquartile Range)

Measures of **Dispersion** tell us **how spread out the data is**.

The two common measures are:

1. **Range**
2. **Interquartile Range (IQR)**

---

# 1. Range

The **Range** is the difference between the **maximum value** and the **minimum value** in a dataset.

### Formula

Range = Maximum Value − Minimum Value

### Example

Data = 10, 20, 30, 40, 50

Range = 50 − 10 = **40**

This means the data is spread over **40 units**.

---

# 2. Interquartile Range (IQR)

The **Interquartile Range (IQR)** measures the spread of the **middle 50% of the data**.

It is calculated using the first quartile (Q1) and third quartile (Q3).

### Formula

IQR = Q3 − Q1

Where:

- **Q1** = 25th Percentile
- **Q2** = 50th Percentile (Median)
- **Q3** = 75th Percentile

### Example

Data = 10, 20, 30, 40, 50, 60, 70, 80, 90

- Q1 = 30
- Q2 (Median) = 50
- Q3 = 70

IQR = 70 − 30 = **40**

This means the middle 50% of the data is spread over **40 units**.

---

# Difference Between Range and IQR

| Range | IQR |
|--------|-----|
| Uses minimum and maximum values | Uses Q1 and Q3 |
| Affected by outliers | Less affected by outliers |
| Measures total spread of data | Measures spread of the middle 50% of data |

---

# Real-Life Example

Suppose the salaries of employees are:

20,000, 22,000, 24,000, 25,000, 26,000, 28,000, 30,000, 32,000, 2,00,000

- **Range** is very large because of the salary **2,00,000** (an outlier).
- **IQR** ignores the extreme value and measures only the spread of the middle salaries.

So, **IQR is more reliable when outliers are present.**

---

# Key Points

- **Range = Maximum Value − Minimum Value**
- **IQR = Q3 − Q1**
- **Q1 = 25th Percentile**
- **Q2 = 50th Percentile (Median)**
- **Q3 = 75th Percentile**
- **Range is affected by outliers.**
- **IQR is less affected by outliers and is a more robust measure of spread.**
```

````python
## code
````

````python
import pandas as pd

## Sample data
data = [10, 20, 30, 40, 50, 60, 70, 80, 90]

df = pd.DataFrame(data, columns=['Values'])

## Calculate Q1 and Q3
Q1 = df['Values'].quantile(0.25)
Q3 = df['Values'].quantile(0.75)

## Calculate IQR
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
````

# Box Plot (Whisker Plot)

A **Box Plot** (also called a **Whisker Plot**) is a graph used to **summarize the distribution of data**.

It helps us understand:

- The **center** of the data (Median)
- The **spread** of the data
- The **presence of outliers**

---

# Parts of a Box Plot

### 1. Box

The **box** represents the **middle 50% of the data**.

- Left edge = **Q1 (25th Percentile)**
- Right edge = **Q3 (75th Percentile)**

The width of the box shows the **Interquartile Range (IQR)**.

---

### 2. Median Line

The line inside the box represents the **Median (Q2 or 50th Percentile)**.

It divides the data into **two equal halves**.

---

### 3. Whiskers

The **whiskers** extend from the box to show the **minimum and maximum values** (excluding outliers).

They represent the spread of the remaining data.

---

### 4. Outliers

Any value that lies **far away from the rest of the data** is called an **Outlier**.

In a box plot, outliers are shown as **individual points (dots)**.

---

# Example

Data:

10, 15, 20, 22, 25, 28, 30, 35, 80

- Q1 = 20
- Median = 25
- Q3 = 30
- 80 is an **Outlier**

The box contains the middle 50% of the data, while the point **80** appears separately as an outlier.

---

# Simple Box Plot

```text
Min     Q1      Median      Q3      Max
 |-------|========|========|-------|
         <------ Box ------->
```

If there is an outlier:

```text
 |-------|========|========|------|      •
                                     Outlier
```

---

# Why is a Box Plot useful?

- Shows the data distribution quickly.
- Helps identify outliers.
- Easy to compare multiple datasets.
- Summarizes the data using five values:
  - Minimum
  - Q1
  - Median
  - Q3
  - Maximum

---

# Key Points

- **Box = Middle 50% of the data (Q1 to Q3).**
- **Line inside the box = Median (Q2).**
- **Whiskers = Minimum and Maximum values (excluding outliers).**
- **Dots outside the whiskers = Outliers.**

````python
import matplotlib.pyplot as plt

# Sample data
data = [10, 15, 20, 22, 25, 28, 30, 35, 80,90]

# Create box plot
plt.boxplot(data)

plt.title("Box Plot")
plt.ylabel("Values")
plt.show()
````

## Measures of Dispersion: Variance and Standard Deviation

# Measures of Dispersion: Variance and Standard Deviation

Variance and Standard Deviation measure **how much the data is spread out from the mean (average).**

If the values are close to the mean, the spread is **small**.

If the values are far from the mean, the spread is **large**.

---

# 1. Variance

**Variance** measures the **average of the squared differences from the mean**.

### Formula

Variance = Average of (Value − Mean)²

### Steps to Calculate Variance

1. Find the mean.
2. Subtract the mean from each value.
3. Square each difference.
4. Find the average of the squared differences.

---

## Example

Data:

10, 20, 30

### Step 1: Find Mean

Mean = (10 + 20 + 30) / 3 = **20**

### Step 2: Difference from Mean

| Value | Difference |
|-------|------------|
| 10 | -10 |
| 20 | 0 |
| 30 | 10 |

### Step 3: Square the Differences

| Difference | Squared |
|------------|---------|
| -10 | 100 |
| 0 | 0 |
| 10 | 100 |

### Step 4: Find Average

Variance = (100 + 0 + 100) / 3

Variance = **66.67**

---

# 2. Standard Deviation

**Standard Deviation** is the **square root of the variance**.

### Formula

Standard Deviation = √Variance

### Example

Variance = 66.67

Standard Deviation = √66.67

≈ **8.16**

---

# Difference Between Variance and Standard Deviation

| Variance | Standard Deviation |
|----------|---------------------|
| Average of squared differences from the mean | Square root of variance |
| Unit is squared | Same unit as the original data |
| Harder to interpret | Easier to understand |

---

# Real-Life Example

Suppose two students have the following marks:

Student A:

80, 81, 79, 80, 80

Student B:

50, 70, 80, 90, 110

Both students have a similar average.

- **Student A** has a **small variance and standard deviation** because the marks are close to the average.
- **Student B** has a **large variance and standard deviation** because the marks are spread out.

---

# Why is Standard Deviation Preferred?

Standard deviation is easier to understand because it is in the **same unit as the data**.

For example:

- Heights → Standard deviation is in **cm**
- Marks → Standard deviation is in **marks**
- Salary → Standard deviation is in **rupees**

Variance is in **squared units**, so it is less intuitive.

---

# Applications

Variance and Standard Deviation are commonly used in:

- Machine Learning
- Data Analysis
- Finance (Stock Market Volatility)
- Risk Analysis
- Quality Control

---

# Key Points

- **Variance = Average of squared differences from the mean.**
- **Standard Deviation = Square root of variance.**
- **Small variance / standard deviation → Data is close to the mean.**
- **Large variance / standard deviation → Data is widely spread.**
- **Standard deviation is easier to interpret because it uses the original unit of the data.**

## Example: Standard Deviation

### Dataset 1

Data:

169, 170, 170, 171, 170

- Mean = 170
- Standard Deviation ≈ 0.63

**Interpretation:**

The values are very close to the mean, so the standard deviation is **small**.

---

### Dataset 2

Data:

150, 160, 170, 180, 190

- Mean = 170
- Standard Deviation ≈ 14.14

**Interpretation:**

The values are far from the mean, so the standard deviation is **large**.

---

### Key Point

- **Small Standard Deviation** → Data is close to the mean.
- **Large Standard Deviation** → Data is spread far from the mean.

## Why is the Standard Deviation Larger?

### Dataset 1

Data:

169, 170, 170, 171, 170

Mean = **170**

| Value | Distance from Mean |
|------:|-------------------:|
| 169 | 1 |
| 170 | 0 |
| 170 | 0 |
| 171 | 1 |
| 170 | 0 |

**Standard Deviation ≈ 0.63**

**Interpretation:**

The values are very close to the mean (170). The maximum distance from the mean is only **1 unit**, so the standard deviation is **small**.

---

### Dataset 2

Data:

150, 160, 170, 180, 190

Mean = **170**

| Value | Distance from Mean |
|------:|-------------------:|
| 150 | 20 |
| 160 | 10 |
| 170 | 0 |
| 180 | 10 |
| 190 | 20 |

**Standard Deviation ≈ 14.14**

**Interpretation:**

The values are much farther from the mean. Some values are **20 units away** from the mean, so the standard deviation is **large**.

---

## Comparison

| Dataset | Mean | Standard Deviation | Interpretation |
|---------|------|-------------------:|---------------|
| 169, 170, 170, 171, 170 | 170 | 0.63 | Values are close to the mean |
| 150, 160, 170, 180, 190 | 170 | 14.14 | Values are far from the mean |

---

## Key Point

- **Small Standard Deviation** → Data points are close to the mean.
- **Large Standard Deviation** → Data points are far from the mean.
- **Standard Deviation tells us the typical (average) distance of data points from the mean.**

````python
import pandas as pd

# Sample data
data = [10, 20, 30]

df = pd.DataFrame(data, columns=["Values"])

# Variance
variance = df["Values"].var(ddof=0)   # Population Variance

# Standard Deviation
std = df["Values"].std(ddof=0)        # Population Standard Deviation

print("Variance:", variance)
print("Standard Deviation:", std)
````
