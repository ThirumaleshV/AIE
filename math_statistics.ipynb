## Descriptive Statistics vs Inferential Statistics

Statistics is divided into two main types:

1. **Descriptive Statistics**
2. **Inferential Statistics**

---

## 1. Descriptive Statistics

Descriptive statistics are used to **summarize and describe the data**.

It tells us **what the data looks like**, but it does **not make predictions**.

### Things studied

- Mean (Average)
- Minimum value
- Maximum value
- Variance
- Standard Deviation

### Purpose

To summarize and understand the data.

### Example

Suppose a class has marks:

`70, 75, 80, 85, 90`

Descriptive statistics can tell us:

- Average = 80
- Minimum = 70
- Maximum = 90

It only describes the given data.

---

## 2. Inferential Statistics

Inferential statistics are used to **make predictions or draw conclusions about a population using a sample**.

Instead of studying everyone, we study a small sample and make decisions about the whole population.

### Things studied

- Hypothesis Testing
- Confidence Interval
- Regression

### Purpose

To make predictions or conclusions about a population.

### Example

A college has **10,000 students**.

Instead of asking all students, we survey **200 students**.

If 180 out of 200 students like online classes, we may conclude that **most students in the college like online classes**.

This is called inferential statistics because we used a sample to make a conclusion about the entire population.

---

## Difference Between Descriptive and Inferential Statistics

| Descriptive Statistics | Inferential Statistics |
|------------------------|-------------------------|
| Describes the data | Makes predictions or conclusions |
| Uses sample or population | Uses sample only |
| Summarizes data | Predicts about the population |
| Example: Mean, Min, Max, Variance | Example: Hypothesis Testing, Confidence Interval, Regression |

---

## Real-Life Example

Imagine a school has **1000 students**.

- If we calculate the **average marks of all 1000 students**, it is **Descriptive Statistics**.
- If we check the marks of **100 students** and predict the performance of all 1000 students, it is **Inferential Statistics**.

---

## Key Points

- **Descriptive Statistics = Describe the data.**
- **Inferential Statistics = Predict or conclude about a population using a sample.**
- Descriptive answers **"What does the data look like?"**
- Inferential answers **"What can we predict from the data?"**

## Mean, Median and Mode

Mean, Median, and Mode are three common ways to find the **center** of a dataset.

---

## 1. Mean (Average)

The **Mean** is the **average** of all values.

### Formula

Mean = (Sum of all values) / (Number of values)

### Example

Marks = 70, 80, 90

Mean = (70 + 80 + 90) / 3

Mean = 240 / 3 = **80**

### When to use?

Use the mean when there are **no extreme (very high or very low) values** in the data.

---

## 2. Median

The **Median** is the **middle value** after arranging the data in ascending order.

### Example (Odd number of values)

Data = 10, 20, 30, 40, 50

Median = **30**

(The middle value)

### Example (Even number of values)

Data = 10, 20, 30, 40

There are two middle values: 20 and 30

Median = (20 + 30) / 2 = **25**

### When to use?

Use the median when the data contains **outliers (extreme values)** because it is not affected much by them.

---

## 3. Mode

The **Mode** is the value that appears **most frequently** in the dataset.

### Example

Data = 2, 3, 3, 4, 5

Mode = **3**

because it occurs **2 times**, which is more than any other number.

### When to use?

Mode is useful for **categorical data** or to find the **most common value**.

Example:
- Favorite color of students
- Most purchased product

---

## Comparison

| Measure | Meaning | Example |
|---------|---------|---------|
| Mean | Average of all values | (10 + 20 + 30) / 3 = 20 |
| Median | Middle value after sorting | 10, 20, 30, 40, 50 → 30 |
| Mode | Most frequently occurring value | 2, 3, 3, 4 → 3 |

---

## Real-Life Example

Suppose the ages of five friends are:

18, 20, 20, 22, 25

- **Mean** = (18 + 20 + 20 + 22 + 25) / 5 = **21**
- **Median** = **20**
- **Mode** = **20**

---

## Key Points

- **Mean = Average**
- **Median = Middle value after sorting**
- **If there are an even number of values, Median = Average of the two middle values**
- **Mode = Most frequently occurring value**

## Percentile

A **Percentile** tells us the position of a value in a dataset.

It shows **what percentage of the data is less than or equal to that value**.

Before finding percentiles, the data must be arranged in **ascending order**.

---

## Example

Scores:

38, 41, 45, 50, 54, 65, 66, 69, 78

---

## 25th Percentile (Q1)

The **25th percentile** is the value below which **25% of the data** falls.

In this example:

**25th Percentile = 45**

This means **25% of the scores are less than or equal to 45**.

---

## 50th Percentile (Median or Q2)

The **50th percentile** is also called the **Median**.

It divides the dataset into **two equal halves**.

In this example:

**50th Percentile = 54**

This means **50% of the scores are less than or equal to 54**.

---

## 75th Percentile (Q3)

The **75th percentile** is the value below which **75% of the data** falls.

In this example:

**75th Percentile = 66**

This means **75% of the scores are less than or equal to 66**.

---

## Real-Life Example

Suppose you scored in the **90th percentile** in an exam.

This means:

- You scored **better than 90% of the students**.
- Only **10% of the students scored higher than you**.

---

## Quartiles

- **25th Percentile = First Quartile (Q1)**
- **50th Percentile = Second Quartile (Q2) = Median**
- **75th Percentile = Third Quartile (Q3)**

---

## Key Points

- Percentile shows the **position** of a value in a dataset.
- Data should be **sorted** before finding percentiles.
- **25th Percentile (Q1)** → 25% of data is below or equal to this value.
- **50th Percentile (Q2)** → Median (middle value).
- **75th Percentile (Q3)** → 75% of data is below or equal to this value.

## Measures of Dispersion: Range and IQR (Interquartile Range)

Measures of **Dispersion** tell us **how spread out the data is**.

The two common measures are:

1. **Range**
2. **Interquartile Range (IQR)**

---

## 1. Range

The **Range** is the difference between the **maximum value** and the **minimum value** in a dataset.

### Formula

Range = Maximum Value − Minimum Value

### Example

Data = 10, 20, 30, 40, 50

Range = 50 − 10 = **40**

This means the data is spread over **40 units**.

---

## 2. Interquartile Range (IQR)

The **Interquartile Range (IQR)** measures the spread of the **middle 50% of the data**.

It is calculated using the first quartile (Q1) and third quartile (Q3).

### Formula

IQR = Q3 − Q1

Where:

- **Q1** = 25th Percentile
- **Q2** = 50th Percentile (Median)
- **Q3** = 75th Percentile

### Example

Data = 10, 20, 30, 40, 50, 60, 70, 80, 90

- Q1 = 30
- Q2 (Median) = 50
- Q3 = 70

IQR = 70 − 30 = **40**

This means the middle 50% of the data is spread over **40 units**.

---

## Difference Between Range and IQR

| Range | IQR |
|--------|-----|
| Uses minimum and maximum values | Uses Q1 and Q3 |
| Affected by outliers | Less affected by outliers |
| Measures total spread of data | Measures spread of the middle 50% of data |

---

## Real-Life Example

Suppose the salaries of employees are:

20,000, 22,000, 24,000, 25,000, 26,000, 28,000, 30,000, 32,000, 2,00,000

- **Range** is very large because of the salary **2,00,000** (an outlier).
- **IQR** ignores the extreme value and measures only the spread of the middle salaries.

So, **IQR is more reliable when outliers are present.**

---

## Key Points

- **Range = Maximum Value − Minimum Value**
- **IQR = Q3 − Q1**
- **Q1 = 25th Percentile**
- **Q2 = 50th Percentile (Median)**
- **Q3 = 75th Percentile**
- **Range is affected by outliers.**
- **IQR is less affected by outliers and is a more robust measure of spread.**
```

````python
## code
````

````python
import pandas as pd

## Sample data
data = [10, 20, 30, 40, 50, 60, 70, 80, 90]

df = pd.DataFrame(data, columns=['Values'])

## Calculate Q1 and Q3
Q1 = df['Values'].quantile(0.25)
Q3 = df['Values'].quantile(0.75)

## Calculate IQR
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
````

## Box Plot (Whisker Plot)

A **Box Plot** (also called a **Whisker Plot**) is a graph used to **summarize the distribution of data**.

It helps us understand:

- The **center** of the data (Median)
- The **spread** of the data
- The **presence of outliers**

---

## Parts of a Box Plot

### 1. Box

The **box** represents the **middle 50% of the data**.

- Left edge = **Q1 (25th Percentile)**
- Right edge = **Q3 (75th Percentile)**

The width of the box shows the **Interquartile Range (IQR)**.

---

### 2. Median Line

The line inside the box represents the **Median (Q2 or 50th Percentile)**.

It divides the data into **two equal halves**.

---

### 3. Whiskers

The **whiskers** extend from the box to show the **minimum and maximum values** (excluding outliers).

They represent the spread of the remaining data.

---

### 4. Outliers

Any value that lies **far away from the rest of the data** is called an **Outlier**.

In a box plot, outliers are shown as **individual points (dots)**.

---

## Example

Data:

10, 15, 20, 22, 25, 28, 30, 35, 80

- Q1 = 20
- Median = 25
- Q3 = 30
- 80 is an **Outlier**

The box contains the middle 50% of the data, while the point **80** appears separately as an outlier.

---

## Simple Box Plot

```text
Min     Q1      Median      Q3      Max
 |-------|========|========|-------|
         <------ Box ------->
```

If there is an outlier:

```text
 |-------|========|========|------|      •
                                     Outlier
```

---

## Why is a Box Plot useful?

- Shows the data distribution quickly.
- Helps identify outliers.
- Easy to compare multiple datasets.
- Summarizes the data using five values:
  - Minimum
  - Q1
  - Median
  - Q3
  - Maximum

---

## Key Points

- **Box = Middle 50% of the data (Q1 to Q3).**
- **Line inside the box = Median (Q2).**
- **Whiskers = Minimum and Maximum values (excluding outliers).**
- **Dots outside the whiskers = Outliers.**

````python
import matplotlib.pyplot as plt

# Sample data
data = [10, 15, 20, 22, 25, 28, 30, 35, 80,90]

# Create box plot
plt.boxplot(data)

plt.title("Box Plot")
plt.ylabel("Values")
plt.show()
````

## Measures of Dispersion: Variance and Standard Deviation

Variance and Standard Deviation measure **how much the data is spread out from the mean (average).**

If the values are close to the mean, the spread is **small**.

If the values are far from the mean, the spread is **large**.

---

## 1. Variance

**Variance** measures the **average of the squared differences from the mean**.

### Formula

Variance = Average of (Value − Mean)²

### Steps to Calculate Variance

1. Find the mean.
2. Subtract the mean from each value.
3. Square each difference.
4. Find the average of the squared differences.

---

## Example

Data:

10, 20, 30

### Step 1: Find Mean

Mean = (10 + 20 + 30) / 3 = **20**

### Step 2: Difference from Mean

| Value | Difference |
|-------|------------|
| 10 | -10 |
| 20 | 0 |
| 30 | 10 |

### Step 3: Square the Differences

| Difference | Squared |
|------------|---------|
| -10 | 100 |
| 0 | 0 |
| 10 | 100 |

### Step 4: Find Average

Variance = (100 + 0 + 100) / 3

Variance = **66.67**

---

## 2. Standard Deviation

**Standard Deviation** is the **square root of the variance**.

### Formula

Standard Deviation = √Variance

### Example

Variance = 66.67

Standard Deviation = √66.67

≈ **8.16**

---

## Difference Between Variance and Standard Deviation

| Variance | Standard Deviation |
|----------|---------------------|
| Average of squared differences from the mean | Square root of variance |
| Unit is squared | Same unit as the original data |
| Harder to interpret | Easier to understand |

---

## Real-Life Example

Suppose two students have the following marks:

Student A:

80, 81, 79, 80, 80

Student B:

50, 70, 80, 90, 110

Both students have a similar average.

- **Student A** has a **small variance and standard deviation** because the marks are close to the average.
- **Student B** has a **large variance and standard deviation** because the marks are spread out.

---

## Why is Standard Deviation Preferred?

Standard deviation is easier to understand because it is in the **same unit as the data**.

For example:

- Heights → Standard deviation is in **cm**
- Marks → Standard deviation is in **marks**
- Salary → Standard deviation is in **rupees**

Variance is in **squared units**, so it is less intuitive.

---

## Applications

Variance and Standard Deviation are commonly used in:

- Machine Learning
- Data Analysis
- Finance (Stock Market Volatility)
- Risk Analysis
- Quality Control

---

## Key Points

- **Variance = Average of squared differences from the mean.**
- **Standard Deviation = Square root of variance.**
- **Small variance / standard deviation → Data is close to the mean.**
- **Large variance / standard deviation → Data is widely spread.**
- **Standard deviation is easier to interpret because it uses the original unit of the data.**

## Example: Standard Deviation

### Dataset 1

Data:

169, 170, 170, 171, 170

- Mean = 170
- Standard Deviation ≈ 0.63

**Interpretation:**

The values are very close to the mean, so the standard deviation is **small**.

---

### Dataset 2

**Interpretation:**

The values are far from the mean, so the standard deviation is **large**.

---

### Key Point

- **Small Standard Deviation** → Data is close to the mean.
- **Large Standard Deviation** → Data is spread far from the mean.

## Why is the Standard Deviation Larger?

### Dataset 1

Data:

169, 170, 170, 171, 170

Mean = **170**

| Value | Distance from Mean |
|------:|-------------------:|
| 169 | 1 |
| 170 | 0 |

**Standard Deviation ≈ 0.63**

**Interpretation:**

---

### Dataset 2

| Value | Distance from Mean |
|------:|-------------------:|
| 150 | 20 |
| 160 | 10 |

---

## Comparison

| Dataset | Mean | Standard Deviation | Interpretation |
|---------|------|-------------------:|---------------|
| 169, 170, 170, 171, 170 | 170 | 0.63 | Values are close to the mean |
| 150, 160, 170, 180, 190 | 170 | 14.14 | Values are far from the mean |

---

## Key Point

- **Small Standard Deviation** → Data points are close to the mean.
- **Large Standard Deviation** → Data points are far from the mean.
- **Standard Deviation tells us the typical (average) distance of data points from the mean.**

## Interpreting Standard Deviation

There is **no fixed standard deviation value** (such as 5 or 10) that always means the data is spread out or not spread out.

The interpretation depends on the **mean** and the **scale of the data**.

## Rule of Thumb

| Standard Deviation Compared to Mean | Conclusion |
|-------------------------------------|------------|
| Much smaller than the mean | Data is **not spread much** |
| Around 10–30% of the mean | Data is **moderately spread** |
| Close to or larger than the mean | Data is **highly spread** |

## Examples

### Example 1

```text
Mean = 100
Standard Deviation = 5
```

**Conclusion:** The data is **not spread much** because most values are close to the mean.

---

### Example 2

```text
Mean = 100
Standard Deviation = 20
```

**Conclusion:** The data is **moderately spread**.

---

### Example 3

```text
Mean = 100
Standard Deviation = 80
```

**Conclusion:** The data is **highly spread** because the values are far from the mean.

---

## Important Note

This is only a **rule of thumb**, not a strict rule.

To judge how spread out the data is, consider:

- The **mean**
- The **range of the data**
- Comparing the **standard deviation** with other datasets that measure the same thing

---

## Example

```text
Mean = 50
Standard Deviation = 7.15
```

**Conclusion:** Since **7.15 is much smaller than 50**, the data is **not spread much**. Most values are fairly close to the mean.

````python
import pandas as pd

# Sample data
data = [10, 20, 30]

df = pd.DataFrame(data, columns=["Values"])

# Variance
variance = df["Values"].var(ddof=0)   # Population Variance

# Standard Deviation
std = df["Values"].std(ddof=0)        # Population Standard Deviation

print("Variance:", variance)
print("Standard Deviation:", std)
````

## Correlation

**Correlation** measures the relationship between **two variables**.

It tells us **how one variable changes when another variable changes**.

---

## Example

| Commute Time (minutes) | Employee Happiness |
|------------------------|-------------------:|
| 0 | 9 |
| 5 | 9 |
| 10 | 8 |
| 25 | 7 |
| 30 | 7 |
| 60 | 4 |
| 120 | 2 |

As **commute time increases**, the **happiness index decreases**.

This is an example of **negative correlation**.

---

## Correlation Coefficient (r)

The correlation coefficient ranges from **-1 to +1**.

| Correlation Value | Meaning |
|------------------:|---------|
| +1 | Perfect Positive Correlation |
| 0 | No Correlation |
| -1 | Perfect Negative Correlation |

---

## Types of Correlation

### 1. Positive Correlation (+1)

As one variable **increases**, the other variable **also increases**.

### Example

| Study Hours | Marks |
|-------------|------:|
| 2 | 40 |
| 4 | 60 |
| 6 | 80 |
| 8 | 95 |

More study hours → Higher marks.

---

### 2. Negative Correlation (-1)

As one variable **increases**, the other variable **decreases**.

### Example

| Commute Time | Happiness |
|-------------:|----------:|
| 10 | 8 |
| 30 | 7 |
| 60 | 4 |
| 120 | 2 |

Longer commute → Lower happiness.

---

### 3. No Correlation (0)

There is **no relationship** between the two variables.

### Example

| Shoe Size | Marks |
|-----------|------:|
| 6 | 85 |
| 8 | 70 |
| 7 | 90 |
| 9 | 60 |

Shoe size has no effect on marks.

---

## Key Points

- Correlation measures the relationship between two variables.
- Correlation value always lies between **-1 and +1**.
- **+1** → Perfect Positive Correlation.
- **0** → No Correlation.
- **-1** → Perfect Negative Correlation.
- **Correlation does not mean causation.**

````python
import pandas as pd

# Sample data
df = pd.DataFrame({
    "Commute_Time": [10, 60, 0, 120, 30, 25, 5],
    "Happiness": [8, 4, 9, 2, 7, 7, 9]
})

# Correlation
corr = df["Commute_Time"].corr(df["Happiness"])

print("Correlation:", corr)
````

## Correlation vs Causation

## Correlation

### Definition
Correlation means **two variables change together**.

It **does not mean** one variable causes the other.

### Easy Example

Suppose on a hot day:

- 🍦 Ice cream sales increase.
- 🕶️ Sunglasses sales also increase.

Both increase together.

Does buying ice cream cause people to buy sunglasses?

**No.**

The real reason is **hot weather**.

So, ice cream sales and sunglasses sales are **correlated**.

---

## Causation

### Definition
Causation means **one variable directly causes another variable to change**.

### Easy Example

Suppose you press a light switch.

- **Cause:** You press the switch.
- **Effect:** The light turns on.

Here, pressing the switch **directly causes** the light to turn on.

This is **causation**.

---

## Difference

| Correlation | Causation |
|-------------|-----------|
| Two variables change together. | One variable directly causes the other. |
| No direct cause-and-effect. | Direct cause-and-effect. |

---

## Key Points

\[

`1, 2, 3, 4, 5, 6`

- Total possible outcomes = **6**

Probability of getting 4:

P(4) = \frac{1}{6}
\]

## Example 1: Tossing a Coin

Possible outcomes:

- Head (H)
- Tail (T)

Total possible outcomes = **2**

What is the probability of getting a Head?

- Favorable outcomes = **1** (Head)
- Total outcomes = **2**

\[
P(\text{Head}) = \frac{1}{2} = 0.5
\]

**Answer:** There is a **50% chance** of getting a Head.

---

## Example 2: Rolling a Dice

Possible outcomes:

`1, 2, 3, 4, 5, 6`

What is the probability of getting an even number?

Even numbers are:

`2, 4, 6`

- Favorable outcomes = **3**
- Total outcomes = **6**

\[
P(\text{Even}) = \frac{3}{6} = \frac{1}{2} = 0.5
\]

**Answer:** The probability is **0.5 (50%)**.

---

## Example 3: Picking a Red Ball

A bag contains:

- 🔴 3 Red balls
- 🔵 2 Blue balls

Total balls = **5**

What is the probability of picking a red ball?

- Favorable outcomes = **3**
- Total outcomes = **5**

\[
P(\text{Red}) = \frac{3}{5} = 0.6
\]

**Answer:** There is a **60% chance** of picking a red ball.

---

## Key Points

- Probability tells us the chance of an event happening.
- Probability is always between **0 and 1**.
- **0** means impossible.
- **1** means certain.

\[
\text{Probability} = \frac{\text{Favorable Outcomes}}{\text{Total Possible Outcomes}}

## Conditional Probability and Bayes' Theorem

## 1. Conditional Probability

### What is Conditional Probability?

Conditional probability means finding the **chance of an event happening when we already know another event has happened**.

In simple words:

> "What is the probability of Event A **given** that Event B has already happened?"

---

## Easy Example

A bag contains:

- 🔴 3 Red balls
- 🔵 2 Blue balls

Total balls = **5**

Suppose someone tells you:

> "The ball picked is **not blue**."

Now only the **red balls** are possible.

So,

Probability of getting a red ball **given that the ball is not blue** = 3/3 = **1**

**Answer:** 100%

Here, we used extra information ("not blue") to find the probability.

---

## Another Easy Example

A class has:

- 6 Boys
- 4 Girls

Total students = **10**

Suppose you already know the selected student is a **boy**.

What is the probability that the student is a boy?

Since only boys are possible now:

Probability = 6/6 = **1 (100%)**

---

## 2. Bayes' Theorem

### What is Bayes' Theorem?

Bayes' Theorem is a formula used to **update the probability** when we get **new information**.

It helps answer questions like:

> "Now that I know this information, what is the new probability?"

---

## Easy Example

Suppose there are **100 students**.

- 60 students play Cricket.
- 40 students play Football.
- Out of the 60 cricket players, 30 also wear glasses.

Now a student is wearing glasses.

What is the chance that the student plays Cricket?

Bayes' Theorem helps answer this type of question.

It uses the new information (**wearing glasses**) to find the updated probability.

---

## Formula

P(E|F) = (P(F|E) × P(E)) / P(F)

Where:

- **P(E|F)** = Probability of E given F
- **P(F|E)** = Probability of F given E
- **P(E)** = Probability of E
- **P(F)** = Probability of F

---

## Difference Between Conditional Probability and Bayes' Theorem

| Conditional Probability | Bayes' Theorem |
|--------------------------|----------------|
| Finds probability using known information. | Updates probability using new information. |
| Uses the condition directly. | Uses a formula to calculate the updated probability. |

---

## Easy Trick to Remember

### Conditional Probability

Think:

> "I already know something."

Example:

"I already know the student is from Class A."

Now find another probability.

---

### Bayes' Theorem

Think:

> "I got new information."

Example:

"I now know the student wears glasses."

Update the probability using this new information.

---

## One-Line Summary

- **Conditional Probability:** Probability after knowing another event.
- **Bayes' Theorem:** A formula used to calculate conditional probability.

## Skewness

## What is Skewness?

**Skewness** tells us whether the data is **balanced or tilted to one side**.

It shows **which side has the longer tail**.

There are **3 types** of skewness:

1. Right-Skewed (Positive Skew)
2. Left-Skewed (Negative Skew)
3. Zero-Skewed (Normal Distribution)

---

## 1. Right-Skewed Distribution (Positive Skew)

## Definition

Most values are **small** and are on the **left**.

Only a few **large values** are on the **right**.

The tail is **longer on the right**.



### Example

Salaries in a company:

```
20k, 22k, 25k, 26k, 28k, 30k, 1,00,000
```

Most employees earn around **20k–30k**.

Only one employee earns **1,00,000**.

So the data is **right-skewed**.

### Relationship

**Mean > Median**

The very high value pulls the **mean** to the right.

---

## 2. Left-Skewed Distribution (Negative Skew)

## Definition

Most values are **large** and are on the **right**.

Only a few **small values** are on the **left**.

The tail is **longer on the left**.



### Example

Marks in an easy exam:

```
30, 85, 88, 90, 92, 95, 98
```

Most students scored **high marks**.

Only one student scored **30**.

So the data is **left-skewed**.

### Relationship

**Median > Mean**

The very low value pulls the **mean** to the left.

---

## 3. Zero-Skewed Distribution (Normal Distribution)

## Definition

Data is **evenly spread** on both sides.

Both sides look the same.



### Example

Marks:

```
45, 50, 55, 60, 65, 70, 75
```

The values are balanced around the center.

So this is a **normal distribution**.

### Relationship

**Mean = Median = Mode**

---

## Difference

| Right Skew | Left Skew | Zero Skew |
|------------|-----------|-----------|
| Tail on the right | Tail on the left | No long tail |
| Most values on the left | Most values on the right | Values are balanced |
| Mean > Median | Median > Mean | Mean = Median = Mode |

---

## Easy Trick to Remember

### Right Skew

👉 Tail points to the **Right**

➡️ **Mean > Median**

---

### Left Skew

👉 Tail points to the **Left**

⬅️ **Median > Mean**

---

### Zero Skew

👉 Perfectly balanced

**Mean = Median = Mode**

---

## One-Line Summary

- **Right Skew** → Tail on the **right**, Mean > Median.
- **Left Skew** → Tail on the **left**, Median > Mean.
- **Zero Skew** → Balanced, Mean = Median = Mode.

## Standard Deviation (σ) and the 68–95–99.7 Rule

A **Normal Distribution** (also called a **Bell Curve**) follows the **Empirical Rule**, which states that:

- **68.3%** of the data lies within **±1σ** (1 standard deviation)
- **95.5%** of the data lies within **±2σ** (2 standard deviations)
- **99.7%** of the data lies within **±3σ** (3 standard deviations)

                                      Mean (μ)
                                         │
                                         │
                                  .-''''''''''-.
                               .-'              '-.
                             .'                    '.
                           .'                        '.
                         .'                            '.
                       .'                                '.
___________________ _.'_ ____ _____________________________'.____________________
      -3σ          -2σ          -1σ      μ             +1σ        +2σ       +3σ


          -3<------------------------- 99.7% --------------------------->+3

                -2<------------------- 95.5% ------------------->+2

                          -1<----------- 68.3% ----------->+1


Outside -3σ                                          Outside +3σ
   Outlier                                               Outlier

Range:

➡️ About **95.5%** of students scored between **50 and 90**.

### 3. Within ±3σ (99.7%)

Approximately **99.7%** of all observations are within **3 standard deviations** from the mean.

Range:

## Z-Score

## Random Sampling & Sample Bias

## Random Sampling

**Random Sampling** means selecting people or items **randomly** from the population.

Everyone has an **equal chance** of being selected.

### Example

A school has **1000 students**.

To conduct a survey, the teacher randomly selects **100 students**.

This is **Random Sampling**.

---

## Sample Bias

**Sample Bias** happens when the selected sample **does not represent the whole population**.

This can lead to **incorrect conclusions**.

### Example

A school has students from **Grades 1 to 10**.

If the teacher surveys **only Grade 10 students**, the sample is **biased** because it does not represent the entire school.

---

## Common Types of Bias

### 1. Participation Bias

Occurs when only people who **choose to participate** are included.

**Example:**

An online survey is filled only by people interested in technology.

---

### 2. Language Bias

Occurs when the wording of a question influences the answer.

**Example:**

Instead of asking:

> "Do you like this phone?"

Asking:

> "Don't you think this phone is amazing?"

The second question may influence people's answers.

---

### 3. Survivorship Bias

Occurs when we **only look at successful cases** and ignore failures.

**Example:**

Only studying successful companies and ignoring companies that failed.

This can lead to wrong conclusions.

---

## Key Points

- **Random Sampling** → Everyone has an equal chance of being selected.
- **Sample Bias** → The sample does not represent the whole population.
- **Participation Bias** → Only volunteers participate.
- **Language Bias** → The wording of a question influences answers.
- **Survivorship Bias** → Looking only at successes and ignoring failures.

## Central Limit Theorem, Sampling Distribution

## Central Limit Theorem (CLT)

The **Central Limit Theorem (CLT)** states that as the **sample size increases**, the **sample means** tend to follow a **normal (bell-shaped) distribution**, even if the original population is not normally distributed.

A sample size of **30 or more** is usually considered enough.

### Example

Suppose students' marks are not normally distributed.

If we take **many random samples of 30 students** and calculate the **mean** of each sample, those sample means will form a **normal distribution**.

---

## Sampling Distribution (Simple Explanation)

A **Sampling Distribution** is the distribution of a statistic (such as the **sample mean**) obtained by taking **many random samples** from the same population.

Instead of looking at the whole population, we:

1. Take a random sample.
2. Calculate its mean.
3. Repeat this many times.
4. Collect all the sample means.

The collection of these sample means is called the **Sampling Distribution**.

---

## Example

Suppose a school has **1000 students**.

Instead of calculating the average marks of all students, we randomly select **30 students**.

### Sample 1

Average Marks = **68**

### Sample 2

Average Marks = **72**

### Sample 3

Average Marks = **70**

### Sample 4

Average Marks = **69**

Now we have these sample means:

68, 72, 70, 69, ...

This collection of sample means is called the **Sampling Distribution**.

---

## Easy Way to Remember

- **Population** → All students.
- **Sample** → 30 students.
- **Sample Mean** → Average marks of those 30 students.
- **Sampling Distribution** → All the sample means collected from many random samples.

## Standard Error

## Standard Error (SE)

**Standard Error (SE)** tells us **how accurate the sample mean is compared to the true population mean**.

- **Small Standard Error** → Sample mean is more accurate.
- **Large Standard Error** → Sample mean is less accurate.

### Formula

**SE = Standard Deviation / √Sample Size**

\[
SE = \frac{\sigma}{\sqrt{n}}
\]

Where:

- **σ** = Standard Deviation
- **n** = Sample Size

---

## Example

Suppose:

- Standard Deviation = **20**
- Sample Size = **100**

SE = 20 / √100

SE = 20 / 10

SE = **2**

**Interpretation:**

The sample mean is expected to differ from the true population mean by about **2 units**.

---

## Easy Way to Remember

- **Standard Deviation** → Spread of the **data**.
- **Standard Error** → Accuracy of the **sample mean**.

---

## Key Points

- Standard Error measures the accuracy of the sample mean.
- Larger sample size → Smaller Standard Error.
- Smaller Standard Error → More reliable estimate of the population mean.

## Confidence Interval in Machine Learning

## Why is it used?

In Machine Learning, a model's performance (such as accuracy or precision) is measured using test data.

A single metric (e.g., 92% accuracy) does not tell us how reliable the model is.

A **Confidence Interval (CI)** provides a range in which the model's true performance is likely to lie.

---

## Example

Suppose you build a **Spam Email Classifier**.

After testing it on 1,000 emails:

```text
Accuracy = 92%
```

This tells us the model correctly classified 92% of the emails.

---

## Using a Confidence Interval

After calculating a 95% confidence interval:

```text
Accuracy = 92%
95% Confidence Interval = (90%, 94%)
```

### Meaning

We are **95% confident** that the model's true accuracy on new, unseen data lies between **90% and 94%**.

---

## Why is this useful?

Instead of reporting only:

```text
Accuracy = 92%
```

we report:

```text
Accuracy = 92%
95% Confidence Interval = (90%, 94%)
```

This tells us how **reliable** the model's accuracy is.

---

## Comparing Two Models

| Model | Accuracy | 95% Confidence Interval |
|--------|----------|-------------------------|
| Model A | 92% | 90% – 94% |
| Model B | 91% | 90.8% – 91.2% |

### Interpretation

- **Model A** has higher average accuracy.
- **Model B** has a narrower confidence interval, meaning its performance is more consistent.

---

## Real-Life Example

Suppose you build a disease prediction model.

Test results:

95% Confidence Interval = (94%, 98%)
```

### Meaning

The model's true accuracy on future patients is likely between **94% and 98%**.

A common method is **Bootstrapping**.

### Steps

1. Train the machine learning model.
2. Randomly sample the test data many times (with replacement).
3. Calculate the accuracy for each sample.
4. Use these accuracy values to calculate the confidence interval.

Example accuracy values:

92.1%
91.5%
92.6%
91.9%
...
```

After many samples:

```text

```

# Filling Missing Values: Mean vs Median vs Mode

## 1. Mean (Average)

### Use when:
- Data is **normally distributed** (symmetric).
- There are **no significant outliers**.
- The column is **numerical**.

### Examples:
- Age
- Height
- Temperature
- Exam Marks (without extreme values)

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

### Example:
Data:
20, 22, 24, 26, NaN

Mean = **23**

Fill NaN → **23**

---

## 2. Median (Middle Value)

### Use when:
- Data is **skewed**.
- There are **outliers**.
- The column is **numerical**.

### Examples:
- Salary
- House Price
- Income
- Medical Expenses

```python
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
```

### Example:
Data:
10, 12, 13, 14, 1000, NaN

Mean = **209.8** ❌ (Affected by outlier)

Median = **13** ✅

Fill NaN → **13**

---

## 3. Mode (Most Frequent Value)

### Use when:
- The column is **categorical**.
- You want to fill missing values with the **most common value**.

### Examples:
- Gender
- City
- Department
- Color

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```

### Example:
Data:
Male

Female

Male

Male

NaN

Mode = **Male**

Fill NaN → **Male**

---

## Quick Decision Table

| Column Type | Data Distribution | Fill With |
|-------------|-------------------|-----------|
| Numerical | Normal (No Outliers) | **Mean** |
| Numerical | Skewed / Has Outliers | **Median** |
| Categorical | Any Distribution | **Mode** |

---

## Steps Before Filling Missing Values

### 1. Check Missing Values
```python
df.isnull().sum()
```

### 2. Check Distribution
```python
df["Salary"].hist()
```

### 3. Check Outliers
```python

```

### 4. Choose the Method
- No Outliers → **Mean**
- Outliers Present → **Median**
- Categorical Column → **Mode**
