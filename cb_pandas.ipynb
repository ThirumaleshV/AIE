## 7.Data Concatenation and Merging

```python
import pandas as pd
india_weather=pd.DataFrame({
  "city":["mumbai","delhi","bangalore"],
  "temperature":[32,45,30],
  "humidity":[80,60,78]
})
india_weather
```

```python
usa_weather=pd.DataFrame({
  "city":["new york","chicago","oriando"],
  "temperature":[21,14,35],
  "humidity":[68,65,75]
})
usa_weather
```

```python
df=pd.concat([india_weather,usa_weather])
df
```

```python
df=pd.concat([india_weather,usa_weather],ignore_index=True)
df
```

```python
df=pd.concat([india_weather,usa_weather],keys=["india","usa"])
df
```

```python
df.loc["india"]
```

```python
df.loc["india"][0:2]
```

```python
temperature_df = pd.DataFrame({
    "city": ["mumbai","delhi","banglore"],
    "temperature": [32,45,30],
}, index=[0,1,2])
temperature_df
```

```python
windspeed_df = pd.DataFrame({
    "city": ["delhi","mumbai"],
    "windspeed": [7,12],
}, index=[1,0])
windspeed_df
```

```python
df=pd.concat([temperature_df,windspeed_df],axis=1)
df
```

```python
## merge
```

```python
import pandas as pd
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","orlando"],
    "humidity": [65,68,75],
})
df2
```

```python
df3 = pd.merge(df1, df2, on="city")
df3
```

```python
## types of merge
```

```python
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35, 38],
})
df1
```

```python
df2 = pd.DataFrame({
    "city": ["chicago","new york","san diego"],
    "humidity": [65,68,71],
})
df2
```

```python
df3=pd.merge(df1,df2,on="city",how="inner")
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="outer")
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="left")
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="right")
df3
```

## 8.Data Visualization Using Matplotlib and Seaborn

```python
import pandas as pd
path=r"Datasets/linechart.xlsx"
df=pd.read_excel(path)
df.head()
```

```python
# line chart
```

```python
from matplotlib import pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(df["Quarter"],df["Fridge"],color="green")
plt.plot(df["Quarter"],df["Dishwasher"],color="red")
plt.plot(df["Quarter"],df["Washing Machine"],color="orange")
plt.title("Appliances Sales")
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.show()
```

```python
from matplotlib import pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(df["Quarter"],df["Fridge"],color="green",label="Fridge")
plt.plot(df["Quarter"],df["Dishwasher"],color="red",label="Dishwasher")
plt.plot(df["Quarter"],df["Washing Machine"],color="orange",label="Washing Machine")
plt.title("Appliances Sales")
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
