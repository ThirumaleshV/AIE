```python
#reading multiple excel sheets
path=r"C:\Users\thiru\OneDrive\Desktop\Notebooks\Datasets\movies_db.xlsx"
df_movies=pd.read_excel(path,'movies')
df_movies.head(3)
```

```python
df_financials=pd.read_excel(path,'financials')
df_financials.head(3)
```

```python
def stand_currency(curr):
  if curr=="$$" or curr=='USD':
    return "USD"
  return curr
df_financials=pd.read_excel(path,'financials',converters={
  'currency':stand_currency
})
df_financials.head(3)
```

```python
df_merge=pd.merge(df_movies ,df_financials ,on="movie_id")
df_merge.head(4)
```
