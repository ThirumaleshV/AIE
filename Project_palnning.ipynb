# Scope of Work Document (SOW)

- The **Scope of Work Document (SOW)** is a key document in the project planning phase.
- The SOW details the **specific tasks, deliverables, and timelines** for a project, ensuring all parties understand expectations and responsibilities.
- Experiment with the parameters to determine the ones that best fit your dataset and use case.
- We are using **JIRA** as our Project Management tool for this project.
- During the **Backlog grooming meeting**, major project objectives will be converted into actionable tasks in JIRA and assigned to team members.
