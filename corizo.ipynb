# Pandas `axis` - When to Use

## ✅ Use `axis` when applying a function to a **DataFrame**

### `axis=0` → Apply to **columns** (default)

```python
df.apply(function, axis=0)
```

**Example:**

```python
df.apply(sum, axis=0)
```

**Explanation:**  
Each **column** is passed to the function.

---

### `axis=1` → Apply to **rows**

```python
df.apply(function, axis=1)
```

**Example:**

```python
df["Total"] = df.apply(lambda row: row["Math"] + row["Science"], axis=1)
```

**Explanation:**  
Each **row** is passed to the function.

---

## ❌ Do NOT use `axis` when applying a function to a **Series (single column)**

```python
df["price"].apply(lambda x: x * 2)
```

or

```python
df["price"].map(lambda x: x * 2)
```

**Explanation:**  
Here, `x` is a **single value**, so `axis` is **not needed**.

---

## 📌 Easy Trick to Remember

| Object | Use `axis`? | Example |
|--------|-------------|---------|
| `df` (DataFrame) | ✅ Yes | `df.apply(..., axis=1)` |
| `df["column"]` (Series) | ❌ No | `df["price"].apply(...)` |

---

## 🎯 Remember

- **`axis=0`** → Columns (Vertical ⬇️)
- **`axis=1`** → Rows (Horizontal ➡️)
- **DataFrame (`df`)** → Use `axis`
- **Series (`df["column"]`)** → Do **NOT** use `axis`

> **Rule of Thumb:**  
> If you're using **`df.apply()`**, you'll usually need `axis`.  
> If you're using **`df["column"].apply()`**, you **don't** need `axis`.

# Pandas `apply()` vs `transform()`

## `apply()`

`apply()` is used to apply a function to each value, row, column, or group. It is flexible and can return a result of any shape.

### Example

```python
df.groupby("cylinders")["hp"].apply(lambda x: x.mean())
```

Output:

```text
cylinders
4    110
6    160
```

### When to use `apply()`

- Custom operations on groups
- Data summarization
- String manipulation
- Complex calculations
- When the output size can be different from the input size

### Example

```python
cars_df["Company"] = cars_df["name"].apply(
    lambda x: x.split()[0]
)
```

Extracts the company name from each car name.

---

## `transform()`

`transform()` applies a function to each group and returns an output with the same number of rows as the original data.

### Example

```python
df.groupby("cylinders")["hp"].transform("mean")
```

Output:

```text
0    110
1    110
2    160
3    160
```

The mean value of each group is repeated for every row in that group.

### When to use `transform()`

- Filling missing values using group statistics
- Feature engineering
- Creating new columns
- Updating existing columns
- When the output must have the same number of rows as the original data

### Example

```python
cars_df["horsepower"] = cars_df.groupby(
    ["cylinders", "model_year"]
)["horsepower"].transform(
    lambda x: x.fillna(x.mean())
)
```

This replaces missing horsepower values with the mean horsepower of cars having the same cylinders and model year.

---

## Key Difference

### `apply()`

Returns one result per group (or a custom-shaped result).

```python
df.groupby("cylinders")["hp"].apply("mean")
```

Output:

```text
4    110
6    160
```

### `transform()`

Returns one result for every original row.

```python
df.groupby("cylinders")["hp"].transform("mean")
```

Output:

```text
110
110
160
160
```

---

## Quick Comparison

| Feature | apply() | transform() |
|----------|----------|-------------|
| Works with GroupBy | ✅ | ✅ |
| Returns same number of rows | ❌ Not always | ✅ Always |
| Good for summaries | ✅ | ❌ |
| Good for updating columns | ❌ | ✅ |
| Flexible output shape | ✅ | ❌ |
| Fill missing values by group | Possible | ✅ Preferred |

---

## Easy Memory Trick

- **apply()** → "Give me a result for each group."
- **transform()** → "Give me a result for each row."

### Rule of Thumb

If assigning back to a column:

```python
df["new_column"] = ...
```

Prefer:

```python
transform()
```

If generating a summary report:

```python
groupby().apply(...)
```

Use:

```python
apply()
```

- `groupby().transform()` returns results aligned to the original rows and usually keeps the original index.

# `apply()` vs `transform()` in Pandas

## 1. `apply()` → Apply a custom function

Use `apply()` when:

- You want to apply a custom function.
- The output can be a **single value, Series, or DataFrame**.
- The number of rows **can change** (especially with `groupby()`).

### Example 1: Apply on a Series

```python
import pandas as pd

df = pd.DataFrame({
    "price": [50, 80, 120]
})

df["double_price"] = df["price"].apply(lambda x: x * 2)

print(df)
```

**Output:**

```text
   price  double_price
0     50           100
1     80           160
2    120           240
```

---

### Example 2: Apply on a DataFrame

```python
df["total"] = df.apply(
    lambda row: row["Math"] + row["Science"],
    axis=1
)
```

**Explanation:**

- `axis=1` passes **one row at a time** to the function.

---

## 2. `transform()` → Returns the same number of rows

Use `transform()` when:

- You want the output to have the **same length** as the original data.
- It is **mostly used with `groupby()`**.
- Every input row gets one output value.

### Example

```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 40000, 45000]
})

df["Dept Avg"] = df.groupby("Department")["Salary"].transform("mean")

print(df)
```

**Output:**

```text
  Department  Salary  Dept Avg
0         IT   50000   55000.0
1         IT   60000   55000.0
2         HR   40000   42500.0
3         HR   45000   42500.0
```

### Notice

- IT average = **55000** appears for **every IT row**.
- HR average = **42500** appears for **every HR row**.
- The number of rows remains **4**.

---

## `apply()` vs `transform()`

| Feature | `apply()` | `transform()` |
|---------|-----------|---------------|
| Works on | Series, DataFrame, GroupBy | Series, GroupBy |
| Output size | Can change | Must stay the same |
| Custom functions | ✅ Yes | ✅ Yes |
| Mostly used with | General operations | `groupby()` |
| Returns | Scalar, Series, DataFrame | Same length as input |

---

## Easy Way to Remember

### ✅ Use `apply()` when:

- Performing custom calculations.
- Applying a function to each row or column.
- The output size can change.

**Example:**

```python
df["Square"] = df["Marks"].apply(lambda x: x**2)
```

---

### ✅ Use `transform()` when:

- You want the result to have the **same number of rows**.
- You are working with `groupby()` and want to add calculated values back to the original DataFrame.

**Example:**

```python
df["Dept Avg"] = df.groupby("Department")["Salary"].transform("mean")
```

---

## Easy Rule

- `groupby().apply()` → Returns **one result per group** (or a custom object).
- `groupby().transform()` → Returns **one result for every original row** (same number of rows).

---

## Remember

- ✅ `transform()` → **Almost always used with `groupby()`**.
- ✅ `apply()` → Can be used with a **Series**, **DataFrame**, or **GroupBy**.

---

## One-line Memory Trick

- **`apply()`** → **"Apply a function."**
- **`transform()`** → **"Transform values but keep the same number of rows."**

# `axis=0` in Pandas

`axis=0` is used when you want to **apply a function to each column** of a DataFrame.

## Think of it like this

- **`axis=0`** → Columns ⬇️ (goes down each column)
- **`axis=1`** → Rows ➡️ (goes across each row)

---

## Example 1: Find the sum of each column

```python
import pandas as pd

df = pd.DataFrame({
    "Math": [80, 90, 70],
    "Science": [85, 95, 75]
})

print(df.apply(sum, axis=0))
```

**Output:**

```text
Math       240
Science    255
dtype: int64
```

**Explanation:**

`sum()` is applied to each **column**.

---

## Example 2: Find the maximum value in each column

```python
df.apply(max, axis=0)
```

**Output:**

```text
Math       90
Science    95
dtype: int64
```

---

## Example 3: Apply a custom function to each column

```python
df.apply(lambda col: col.mean(), axis=0)
```

**Output:**

```text
Math       80.0
Science    85.0
dtype: float64
```

**Explanation:**

Here, `col` represents an entire **column**.

---

## Difference between `axis=0` and `axis=1`

Suppose the DataFrame is:

```text
   Math  Science
0    80       85
1    90       95
2    70       75
```

## `axis=0`

```python
df.apply(sum, axis=0)
```

**Result:**

```text
Math       240
Science    255
```

**Explanation:**

Each **column** is summed.

---

## `axis=1`

```python
df.apply(sum, axis=1)
```

**Result:**

```text
0    165
1    185
2    145
```

**Explanation:**

Each **row** is summed.

---

## Easy Trick to Remember

- **`axis=0`** → Column-wise → One **column** is passed to the function.
- **`axis=1`** → Row-wise → One **row** is passed to the function.

---

## When do we usually use `axis=0`?

- Find the sum of each column.
- Find the mean of each column.
- Find the maximum or minimum of each column.
- Apply the same function independently to every column.

---

## Note

In practice, **`axis=1`** is used more often because many tasks involve combining values from multiple columns in the same row (for example, `Math + Science`, categorizing rows, creating new columns, etc.).

````python
import pandas as pd

# 1. Create Dataset
data = {
    "Name": ["Arjun", "Yash", "Sana", "Ishaan", "Om"],
    "Maths": [80, 65, 90, 45, 75],
    "Science": [85, 70, 95, 50, 80],
    "English": [75, 60, 85, 55, 70],
    "Study_Hours": [5, 3, 7, 2, 4],
    "Attendance": [90, 75, 95, 60, 80]
}

df = pd.DataFrame(data)

print("Original Data:")
display(df)

# 2. Create Total Marks
df["Total_Marks"] = df[["Maths", "Science", "English"]].sum(axis=1)

# 3. Create Average Marks
df["Average_Marks"] = df[["Maths", "Science", "English"]].mean(axis=1)

# 4. Create Pass/Fail Column
df["Result"] = df["Average_Marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

# 5. Create Attendance Category
df["Attendance_Category"] = df["Attendance"].apply(
    lambda x: "Good" if x >= 75 else "Poor"
)

# 6. Create Study Category
df["Study_Category"] = pd.cut(
    df["Study_Hours"],
    bins=[-1, 2, 4, float("inf")],
    labels=["Low", "Medium", "High"]
)

# 7. Display Final Data
print("\nData After Feature Engineering:")
display(df)
````
