```python
def fact(x):
    if x==0 or x==1:
        return 1
    else:
        return x * fact(x-1)
    
en=int(input("Enter the number:"))
if en <= 0:
    print(f"{en}: this number is not allowed,typer some other number")
else:
    print(f"The factorial of {en} is:{fact(en)}")    
    
# print(fact(4))
```
