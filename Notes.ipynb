# Machine Learning

**Machine Learning (ML)** is a branch of **Artificial Intelligence (AI)** that allows computers to learn from data and make **predictions, decisions, or find patterns** without being explicitly programmed for every task.

## Types of Machine Learning

### 1. Supervised Learning
- Learns from **labeled data**.
- Used for **prediction and classification**.
- Examples: Spam detection, house price prediction, disease classification.

### 2. Unsupervised Learning
- Learns from **unlabeled data**.
- Used to **find patterns, groups, or hidden relationships** in data.
- Examples: Customer segmentation, grouping similar products.

### 3. Reinforcement Learning
- Learns through **rewards and penalties** by interacting with an environment.
- Used for **decision-making and learning optimal actions**.
- Examples: Game AI, robotics, autonomous systems.

### 4. Semi-Supervised Learning
- Uses a combination of **labeled and unlabeled data**.
- Used when getting labeled data is expensive or time-consuming.
- Examples: Image classification and document classification.

### 5. Self-Supervised Learning
- Learns from **unlabeled data by creating its own learning tasks**.
- Used for learning useful representations from large amounts of data.
- Examples: Large Language Models (LLMs), Generative AI, and Computer Vision.

# Common Mistakes in Machine Learning Projects

### 1. Technical Mistake
Using only the **R² score** to decide whether a model is good enough for deployment.

- R² shows how well the model explains the variation in the target.
- It is not enough to measure the complete model performance.
- We should also check other metrics like **MAE, MSE, and RMSE**.

### 2. Management Mistake
Failing to manage the **expectations of project stakeholders**.

- Don't promise that the model will work perfectly.
- Clearly explain the model's strengths and limitations.
- Discuss the results with senior team members before presenting them to business users.

### Key Lesson
**Don't depend on only one metric, and always communicate the model's limitations clearly to stakeholders.**

# About machine learning

## Important Machine Learning Concepts

## 1. Error Analysis
Error analysis is the process of analyzing a model's mistakes to understand why it makes incorrect predictions.

**Example:** Actual house price = ₹50 lakh, Predicted price = ₹65 lakh.
Error = Actual - Predicted = 50 - 65 = -15 lakh.

**Purpose:**
- Identify where the model makes mistakes.
- Find patterns in prediction errors.
- Improve model performance.

## 2. Overfitting and Underfitting

**Overfitting:** The model performs very well on training data but poorly on unseen data. It learns the training data too closely.

**Underfitting:** The model performs poorly on both training and test data because it has not learned the patterns properly.

**Solutions:**
- Overfitting: Use regularization, cross-validation, or a simpler model.
- Underfitting: Use a more suitable or complex model or improve the features.

## 3. Bias and Variance

**Bias:** Error caused by a model being too simple to learn the underlying patterns. High bias can lead to underfitting.

**Variance:** How sensitive a model is to changes in training data. High variance can lead to overfitting.

**Goal:** Find a balance between bias and variance.

## 4. Cross-Validation

Cross-validation is a technique used to evaluate how well a model may perform on unseen data.

**Example:** In 5-fold cross-validation, the dataset is divided into 5 parts. The model trains on 4 parts and tests on the remaining part. This process repeats 5 times, using a different part for testing each time.

**Purpose:**
- Evaluate model performance more reliably.
- Reduce dependence on a single train-test split.
- Help compare different models.

## 5. Feature Engineering

Feature engineering is the process of creating or transforming input features to help a machine learning model learn better.

**Example:** Creating an `Age` column from a `Date of Birth` column.

Other examples:
- Extracting month and year from a date.
- Converting categorical values into numerical values.
- Creating a total-price feature from quantity and unit price.

## 6. Feature Selection

Feature selection means choosing the most useful input features for training a model.

**Example:** To predict house prices, useful features may include location, area, and number of bedrooms.

**Benefits:**
- Reduces unnecessary features.
- Can improve model performance.
- Makes models easier to interpret.
- Can reduce training time.

## 7. Data Leakage

Data leakage occurs when information that would not legitimately be available at prediction time is used during model training.

**Example:** Using a feature that directly reveals the target value.

**Prevention:**
- Split data before fitting preprocessing steps.
- Fit scalers and imputers only on training data.
- Use pipelines to prevent leakage.
- Ensure features contain no information from the future.

## 8. Hyperparameter Tuning

Hyperparameter tuning is the process of finding suitable settings for a machine learning model.

**Examples:**
- `n_estimators` in Random Forest.
- `max_depth` in Decision Tree.
- `C` in SVM.
- `n_neighbors` in KNN.

**Common techniques:**
- GridSearchCV
- RandomizedSearchCV

**Goal:** Find hyperparameters that produce good model performance on validation data.

## 9. Residual Analysis

A residual is the difference between the actual value and the predicted value.

**Formula:**

Residual = Actual Value - Predicted Value

**Example:**
Actual price = ₹40 lakh  
Predicted price = ₹35 lakh  
Residual = ₹5 lakh

Residual analysis helps identify patterns that the model has failed to learn.

A good regression model should generally have residuals scattered around zero without obvious patterns.

## 10. Data Preprocessing

Data preprocessing is the process of preparing raw data before training a machine learning model.

**Common steps:**
- Handle missing values.
- Remove duplicate records when appropriate.
- Handle outliers when necessary.
- Encode categorical variables.
- Scale numerical features when required.
- Split data into training and testing sets.

**Goal:** Make the data suitable for machine learning.

## 11. Model Evaluation

Model evaluation measures how well a trained model performs.

**For Regression:**
- MAE: Mean Absolute Error.
- MSE: Mean Squared Error.
- RMSE: Root Mean Squared Error.
- R² Score: Measures performance relative to predicting the target mean.

**For Classification:**
- Accuracy: Overall proportion of correct predictions.
- Precision: Proportion of predicted positives that are correct.
- Recall: Proportion of actual positives correctly identified.
- F1 Score: Harmonic mean of precision and recall.

The choice of metric depends on the problem and business objective.

## 12. Regularization

Regularization adds a penalty to a model's objective function to discourage overly complex models.

**Types:**

**L1 Regularization (Lasso):**
- Can make some coefficients exactly zero.
- Can help with feature selection.

**L2 Regularization (Ridge):**
- Shrinks coefficients toward zero.
- Usually does not make coefficients exactly zero.

**Goal:** Help reduce overfitting.

## 13. Class Imbalance

Class imbalance occurs when one class contains significantly more examples than another.

**Example:**
- 950 normal transactions.
- 50 fraudulent transactions.

A model that predicts every transaction as normal achieves 95% accuracy but detects no fraud.

**Possible solutions:**
- Use precision, recall, and F1 score.
- Adjust the classification threshold.
- Use class weights.
- Try resampling techniques such as SMOTE when appropriate.

## 14. Model Deployment

Model deployment means making a trained machine learning model available for practical use.

**Example:** Deploying a house-price prediction model through a web application.

**Common tools:**
- FastAPI
- Flask
- Streamlit
- Docker
- Cloud platforms

**Typical process:**
1. Train the model.
2. Evaluate the model.
3. Save the trained model.
4. Create an API or application.
5. Deploy it.
6. Monitor its performance.

## 15. Model Monitoring

Model monitoring means checking whether a deployed model continues to perform well over time.

**Why is it important?**
- Data patterns may change.
- Prediction quality may decrease.
- New types of errors may appear.
- The model may need retraining.

**Example:** A house-price model trained on older market data may become less accurate when property prices change.
