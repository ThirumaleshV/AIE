# Data Drift and Concept Drift

## 1. Data Drift

Data drift occurs when the distribution of input data (features) changes over time.

**Example:** A house-price model was trained using houses mostly between 500 and 1,000 sq. ft. Later, users start entering houses between 2,000 and 4,000 sq. ft.

The input data distribution has changed.

## 2. Concept Drift

Concept drift occurs when the relationship between input features (X) and the target (Y) changes over time.

**Example:** Earlier, house prices depended mainly on location. Later, due to changes in the market, house prices depend more on property age and demand.

The relationship between features and target has changed.

## Difference Between Data Drift and Concept Drift

| Data Drift | Concept Drift |
|---|---|
| Input data distribution changes | Relationship between X and Y changes |
| P(X) changes | P(Y \| X) changes |
| Example: Customer age distribution changes | Example: Customer purchasing behavior changes |

## Solutions for Data Drift and Concept Drift

The main solutions are **Model Monitoring and Retraining**.

## 1. Solutions for Data Drift

- Monitor incoming data.
- Compare new data with training data.
- Check changes in mean, standard deviation, and distribution.
- Retrain the model using updated data when necessary.

## 2. Solutions for Concept Drift

- Monitor model performance over time.
- Collect new data with actual target values.
- Retrain the model to learn updated relationships.
- Evaluate the new model before deployment.

## Model Monitoring and Retraining Process

**Monitor → Detect Drift → Collect New Data → Retrain → Evaluate → Deploy**

## Important Points

- Data Drift → Input data changes.
- Concept Drift → Relationship between input and target changes.
- Both can reduce model performance.
- Monitoring helps detect these changes.
- Retraining may be required when drift affects model performance.
- Do not retrain automatically just because drift is detected. First check whether it actually affects model performance.

# F-Score in Machine Learning

## 1. What is F-Score?

F-Score is a classification evaluation metric that combines Precision and Recall into a single score.

It is useful when accuracy alone is not sufficient, especially for imbalanced datasets.

## 2. Types of F-Score

### A. F1 Score
F1 Score is the harmonic mean of Precision and Recall.

It gives equal importance to both Precision and Recall.

**Use:** When both false positives and false negatives are important.

### B. F0.5 Score
F0.5 Score gives more importance to Precision than Recall.

**Use:** When false positives are costly.

**Example:** Spam detection — incorrectly marking an important email as spam.

### C. F2 Score
F2 Score gives more importance to Recall than Precision.

**Use:** When false negatives are costly.

**Example:** Disease detection — missing a patient with a disease can be dangerous.

## 3. Comparison

| Metric | Importance |
|---|---|
| F0.5 Score | Precision |
| F1 Score | Equal Precision and Recall |
| F2 Score | Recall |

## 4. Important Points

- F1 Score balances Precision and Recall.
- F0.5 prioritizes Precision.
- F2 prioritizes Recall.
- F-Score ranges from 0 to 1.
- Higher F-Score generally indicates better performance.
- F-Score is useful for evaluating classification models, especially with imbalanced datasets.

**Remember:** F0.5 = Precision, F1 = Balance, F2 = Recall.
