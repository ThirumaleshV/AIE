# 10 Stages of AI/ML Project Life Cycle

1. **Requirements** – Understand the business problem and what needs to be solved.

2. **Data Collection** – Collect the required data.

3. **Data Preparation** – Clean and prepare the data for the model.

4. **Exploratory Data Analysis (EDA)** – Explore the data to find patterns, relationships, and problems.

5. **Feature Engineering** – Create or select useful features for the model.

6. **Model Selection & Training** – Choose a suitable ML model and train it.

7. **Model Evaluation** – Check how well the model performs.

8. **Model Fine-Tuning** – Improve the model by adjusting its parameters/settings.

9. **Model Deployment** – Put the trained model into real-world use.

10. **Monitoring & Feedback Loop** – Monitor the model's performance and use feedback to improve it.


# Requirements and Scope of Work (SOW)

## Requirements, SOW & Project Management

## 1. Requirements

**Requirements** define **what the business wants to achieve** from the AI/ML project.

Example: AtliQ Bank wants to develop a **fraud detection system** for Gold credit card transactions in India. :contentReference[oaicite:0]{index=0}

- Understand the business problem.
- Define project goals.
- Identify what is needed to solve the problem.
- Understand the expectations of stakeholders.

## 2. Scope of Work (SOW)

**SOW (Scope of Work)** is a document that clearly defines **what will be done and what will not be done** in the project.

### In Scope
- Data Collection
- Data Preparation
- EDA
- Model Training and Validation
- Model Deployment
- API Development :contentReference[oaicite:1]{index=1}

### Out of Scope
- Platinum credit card transactions
- Transactions from other countries :contentReference[oaicite:2]{index=2}

## 3. Project Management

**Project Management** means **planning, organizing, and tracking the project** so that it is completed successfully.

It includes:

- **Planning** – Decide what needs to be done.
- **Task Management** – Divide the project into smaller tasks.
- **Time Estimation** – Estimate how long each task will take.
- **Team Management** – Coordinate between business and technical teams.
- **Progress Tracking** – Check whether tasks are completed on time.
- **Risk Management** – Identify and handle possible problems.
- **Communication** – Keep stakeholders updated.
- **Tools** – Tools like **Jira, Confluence, Notion, Asana, and Excel** can be used to manage projects.

## 4. Project Estimation

At the beginning of a project, requirements may be **unclear or change over time**.

Therefore, initial time estimates may not be very accurate. As the project progresses and requirements become clearer, the estimates become more accurate.

# Data Collection

## Data Collection

**Data Collection** means gathering the data needed for an AI/ML project.

- Data is usually collected from **inside the organization**.
- Sometimes, companies **buy additional data from external vendors**.
- Data can be extracted from **websites**.
- Python libraries like **Beautiful Soup** can be used for web data extraction.
- Ready-made software such as **Bright Data** can also be used.

## Main Sources of Data

**Internal Data → External/Vendor Data → Web Data**

# Data Cleaning & Exploratory Data Analysis

## Data Cleaning & Exploratory Data Analysis (EDA)

## 1. Data Cleaning

**Data Cleaning** means cleaning and transforming raw data into a format that can be analyzed and used for ML.

Common tasks:

- Handle **missing values**
- Remove **duplicate data**
- Handle **outliers**
- Fix incorrect or inconsistent data
- Encode **categorical variables**
- Normalize or scale features when required

## 2. Exploratory Data Analysis (EDA)

**EDA** means exploring the data to understand its **quality, characteristics, and patterns** before building the model.

Common EDA tasks:

- Check rows and columns
- Check data types
- Find missing values
- Find duplicates and outliers
- Calculate statistics such as mean, median, minimum, and maximum
- Find relationships between features
- Create visualizations such as histograms, box plots, and scatter plots

### Easy Difference

**Data Cleaning → Fix the data**

**EDA → Understand the data**

---

## 3. GDPR

**GDPR (General Data Protection Regulation)** is a data privacy regulation designed to protect people's personal data.

It is especially important when an AI/ML project uses **personal information**.

Examples of personal data:

- Name
- Email
- Phone number
- Address
- Location
- IP address
- Other information that can identify a person

### Important GDPR Principles

- **Data Minimization** → Collect only the data that is necessary.
- **Purpose Limitation** → Use data only for the intended purpose.
- **Data Protection** → Keep personal data secure.
- **Transparency** → People should know how their data is being used.
- **Retention** → Do not keep data longer than necessary.
- **User Rights** → Individuals have rights regarding their personal data.

### In an AI/ML Project

**Collect Data → Protect Personal Data → Clean Data → EDA → Train Model**

# Feature Engineering

## Feature Engineering

**Feature Engineering** is the process of creating or transforming features from existing data so that the ML model can make **better predictions**.

### Example

Suppose we have:

| Experience (Years) | Salary |
|---:|---:|
| 2 | 30,000 |
| 5 | 60,000 |
| 10 | 1,20,000 |

We can create a new feature:

**Salary per Year = Salary / Experience**

This new feature may help the model understand the data better.

### Common Feature Engineering Techniques

- **Create new features** from existing features.
- **Combine features**.
- **Extract useful information** from dates, text, etc.
- **Convert categorical data** into numerical form.
- **Transform numerical features** when needed.

### Simple Example

**Existing Features:**

`Date of Birth → 15-08-2000`

**New Feature:**

`Age → 26`

Instead of giving the model the date directly, we create a more useful feature: **Age**.

### Easy Memory

**Feature Engineering = Create useful features from existing data**

**Better Features → Better information for the model → Better predictions**
