# K Means Clustering: Theory

## What is K-Means?

**K-Means Clustering** is an **unsupervised machine learning algorithm** used to divide data into **K groups (clusters)**.

Each cluster has a central point called a **centroid**.

The algorithm assigns each data point to the **nearest centroid**.

---

## Example

Suppose we have customer data based on:

- Age
- Spending Score

If we want **2 groups**, then:

**K = 2**

The algorithm will create 2 clusters and find the center of each cluster.

### Simple Diagram

                         Cluster 1
                    ●   ●   ●
                  ●   ●  ★  ●
                    ●   ●

                             

                         Cluster 2
                    ●   ●
                  ●   ★   ●
                    ●   ●   ●

                    ★ = Centroid

Each point is assigned to the **nearest centroid**.

---

## How K-Means Works

## Step 1: Choose K

First, decide how many clusters we want.

Example:

**K = 2**

This means we want **2 clusters**.

---

## Step 2: Initialize Centroids

Place **K centroids randomly**.

For K = 2:

**⭐ Centroid 1**

**⭐ Centroid 2**

---

## Step 3: Calculate Distance

Calculate the distance between every data point and each centroid.

Example:

**Point A → Centroid 1 = 2 units**

**Point A → Centroid 2 = 8 units**

Centroid 1 is closer.

Therefore:

**Point A → Cluster 1**

---

## Step 4: Assign Data Points

Every data point is assigned to the **nearest centroid**.

                         ⭐ C1
                       ● ● ●
                     ● ● ●

                                  ● ●
                                ● ● ⭐ C2
                                  ● ●

---

## Step 5: Update Centroids

After assigning the points, calculate the **mean of the points in each cluster**.

The mean becomes the **new centroid**.

**Old Centroid**

↓

**Calculate mean of cluster points**

↓

**New Centroid**

---

## Step 6: Reassign Points

Again calculate the distance between every point and the **new centroids**.

If another centroid is now closer, the point can move to another cluster.

---

## Step 7: Repeat

K-Means repeats these steps:

**Assign points → Update centroids → Assign points → Update centroids**

It continues until the clusters **stop changing** or the algorithm reaches its stopping condition.

---

## K-Means Process

**Choose K**

↓

**Initialize K centroids**

↓

**Calculate distances**

↓

**Assign points to nearest centroid**

↓

**Calculate new centroids**

↓

**Reassign points**

↓

**Repeat**

↓

**Final Clusters**

---

## SSE - Sum of Squared Errors

**SSE = Sum of Squared Errors**

SSE measures the **sum of squared distances between each data point and its cluster's centroid**.

### Formula

**SSE = Σ(distance between point and its centroid)²**

A lower SSE means the points are generally **closer to their centroids**.

---

## Elbow Method

The **Elbow Method** is used to find a suitable value of **K**.

We calculate SSE for different values of K.

| K | SSE |
|---:|---:|
| 1 | 500 |
| 2 | 300 |
| 3 | 180 |
| 4 | 170 |
| 5 | 165 |

As K increases, SSE decreases.

We look for the point where the decrease in SSE **starts to level off**.

This point is called the **Elbow Point**.

### Diagram

│ ●
│   \

│        \
│         ●────●────●
│

          ↑
      Elbow Point

---

## Important Terms

| Term | Meaning |
|---|---|
| **K** | Number of clusters |
| **Cluster** | Group of similar data points |
| **Centroid** | Center point of a cluster |
| **Distance** | Used to find the nearest centroid |
| **SSE** | Sum of squared distances from points to their centroids |
| **Elbow Method** | Method used to choose a suitable K |

---

## Easy Memory Trick

**K-Means = Group similar points around their nearest center.**

**K → Number of clusters**

**Centroid → Center of cluster**

**Distance → Finds nearest centroid**

**SSE → Measures how close points are to their centroids**

**Elbow Method → Helps choose K**

# K Means Clustering: Customer Segmentatio

## K-Means: Important Takeaways

### 1. Feature Scaling

K-Means uses **distance** to assign data points to clusters.

If features have very different scales, one feature can dominate the distance calculation.

Therefore, it is recommended to scale the data using:

- **Min-Max Scaling**
- **Standard Scaling**

Example:

Age → 20–60  
Salary → ₹20,000–₹2,00,000

Salary has a much larger scale, so it can affect the distance more.

**Solution:** Apply feature scaling before K-Means.

---

### 2. Choosing the Value of K

In real-world datasets, there may be many features, so it can be difficult to decide the correct value of **K**.

We can calculate **SSE for different values of K** and plot an **Elbow Chart**.

The point where the decrease in SSE starts to level off is called the **Elbow Point**.

This helps us choose a suitable value of **K**.

---

### 3. Inertia

In K-Means, **inertia** represents the **sum of squared distances between data points and their cluster centroids**.

**Inertia = SSE**

Lower inertia means the data points are closer to their cluster centroids.

In Python:

`kmeans.inertia_`

### Easy Memory Trick

**Different scales → Scale the data**

**Don't know K → Use Elbow Method**

**SSE in K-Means → Inertia**

````python
import pandas as pd
from matplotlib import pyplot as plt

df=pd.read_csv(r"..\Datasets\income.csv")
df.head()
````

````python
plt.scatter(df["Age"],df["Income($)"])
plt.show()
````

````python
from sklearn.cluster import KMeans
km=KMeans(n_clusters=3,n_init="auto") # n_init="auto" tells KMeans to automatically determine how many times to initialize the centroids and select the best clustering result.
km.fit(df[["Age","Income($)"]])
````

````python
km.labels_
````

````python
df["Clusters"]=km.labels_
df.head()
````

````python
import seaborn as sns

sns.scatterplot(df,x='Age',y="Income($)",palette='viridis',hue="Clusters")
````

### Min Max Scaling

Clusters are not formed properly. It can be due to Age and Income columns being on a different scale. We can try scaling them

````python
from sklearn.preprocessing import MinMaxScaler

scaler=MinMaxScaler()

df_scaled=pd.DataFrame(columns=['Age', 'Income($)'])
df_scaled[['Age', 'Income($)']]=scaler.fit_transform(df[["Age","Income($)"]])
````

````python
df_scaled
````

````python
km = KMeans(n_clusters=3, n_init='auto')
km.fit(df_scaled[["Age", "Income($)"]])
df_scaled['Cluster'] = km.labels_
df_scaled.head()
````

````python
sns.scatterplot(df_scaled,x='Age',y="Income($)",palette='viridis',hue="Cluster")
````

````python
centroids=km.cluster_centers_
centroids
````

````python
sns.scatterplot(df_scaled,x='Age',y="Income($)",palette='viridis',hue="Cluster")
sns.scatterplot(x=centroids[:,0], y=centroids[:,1],  s=200, color="red", marker='*')
plt.show()
````

### Elbow method to determine optimal number of clusters

````python
sse=[]
k_rng=range(1,10)

for k in k_rng:
    km=KMeans(n_clusters=k,n_init="auto")
    km.fit(df_scaled[["Age", "Income($)"]])
    sse.append(km.inertia_)

sse    
````

````python
plt.xlabel('K')
plt.ylabel('Sum of squared error')
plt.plot(k_rng,sse)
plt.show()
````

# Hierarchical Clustering Theory

## What is Hierarchical Clustering?

Hierarchical clustering is an **unsupervised learning technique** that groups similar data points into clusters.

It starts with **each data point as its own cluster** and gradually **merges the closest clusters**.

### Steps

1. Treat each data point as its own cluster.
2. Measure the distance between clusters.
3. Merge the closest clusters.
4. Recalculate the distances.
5. Repeat until all data points form one cluster.
6. A **dendrogram** can be used to show the merging process.

### Simple Diagram

<pre>
A   B          C   D

●   ●          ●   ●
│              │
└── Merge ─────┘

(A,B)          (C,D)

       ↓ Merge

       (A,B,C,D)
</pre>

---

## Linkage Methods

**Linkage** tells us how to calculate the distance between two clusters.

There are four important linkage methods:

1. Average Linkage
2. Single Linkage
3. Complete Linkage
4. Ward's Linkage

---

## 1. Average Linkage

Average Linkage calculates the **average distance between all pairs of points** from two clusters.

<pre>
Cluster A                 Cluster B

  ●                         ●
  ● -----------↔----------- ●
  ●                         ●

Calculate all pair distances
             ↓
       Take the average
</pre>

### Example

Suppose the distances between points are:

4, 6, 8, 10

Average distance:

(4 + 6 + 8 + 10) / 4 = 7

Therefore:

**Average Linkage = 7**

**Memory Trick:**  
Average → **Average distance**

---

## 2. Single Linkage

Single Linkage uses the **shortest distance** between any two points from the two clusters.

<pre>
Cluster A                 Cluster B

  ●                         ●
  ● -----------↔----------- ●
  ●                         ●
              ↑
        Shortest distance
</pre>

### Example

Suppose the distances are:

4, 6, 8, 10

The shortest distance is:

**4**

Therefore:

**Single Linkage = 4**

**Memory Trick:**  
Single → **Shortest distance**

---

## 3. Complete Linkage

Complete Linkage uses the **longest distance** between any two points from the two clusters.

<pre>
Cluster A                 Cluster B

  ● ======================= ●
  ●                         ●
  ●                         ●
              ↑
         Longest distance
</pre>

### Example

Suppose the distances are:

4, 6, 8, 10

The longest distance is:

**10**

Therefore:

**Complete Linkage = 10**

**Memory Trick:**  
Complete → **Longest distance**

---

## 4. Ward's Linkage

Ward's Linkage chooses the merge that causes the **smallest increase in within-cluster variance**.

<pre>
Cluster A                 Cluster B

  ● ●                       ● ●
   ●                         ●

            ↓ Merge

       ● ● ● ● ● ●

            ↓

Choose the merge that causes
the smallest increase in variance.
</pre>

### Simple Meaning

Ward's method asks:

**"If I merge these two clusters, how much will the variation inside the cluster increase?"**

It chooses the merge with the **minimum increase in variance**.

**Memory Trick:**  
Ward → **Variance**

---

## Linkage Comparison

| Linkage | How it calculates distance |
|---|---|
| **Single Linkage** | Shortest distance |
| **Complete Linkage** | Longest distance |
| **Average Linkage** | Average distance |
| **Ward's Linkage** | Minimum increase in within-cluster variance |

## Easy Memory Trick

- **Single → Shortest**
- **Complete → Longest**
- **Average → Average**
- **Ward → Variance**

---

## Final Takeaway

**Hierarchical Clustering:**

Data points → Individual clusters → Find closest clusters → Merge → Recalculate distance → Repeat

**Linkage = The rule used to decide how the distance between two clusters is calculated.**

## K-Means vs Agglomerative Clustering

Both **K-Means** and **Agglomerative Clustering** are **unsupervised learning algorithms** used to group similar data points into clusters.

---

## 1. K-Means Clustering

K-Means creates clusters using **centroids**.

### Steps

1. Choose the number of clusters `K`.
2. Select initial centroids.
3. Assign each data point to the nearest centroid.
4. Recalculate the centroids.
5. Repeat until the clusters stop changing.

### Simple Flow

Data Points → Choose K Centroids → Assign Points → Update Centroids → Repeat

### Example

```python
from sklearn.cluster import KMeans

km = KMeans(n_clusters=3)

km.fit(df[["age", "income"]])

df["cluster"] = km.labels_
```

---

## 2. Agglomerative Clustering

Agglomerative Clustering creates clusters by **merging the closest clusters**.

### Steps

1. Start with every data point as its own cluster.
2. Find the closest clusters.
3. Merge them.
4. Recalculate distances.
5. Repeat until the desired number of clusters is obtained.

### Simple Flow

Each Point = Separate Cluster → Find Closest Clusters → Merge → Repeat

### Example

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(
    n_clusters=3,
    linkage="average"
)

hc.fit(df[["age", "income"]])

df["cluster"] = hc.labels_
```

---

## Main Difference

| K-Means | Agglomerative Clustering |
|---|---|
| Uses **centroids** | Uses **distances between clusters** |
| Assigns points to nearest centroid | Merges closest clusters |
| Centroids are repeatedly updated | Clusters are repeatedly merged |
| Usually uses **Elbow Method** to find K | Usually uses **Dendrogram** to decide clusters |
| Generally faster for large datasets | Can be slower for large datasets |
| Centroid-based | Hierarchical / merge-based |

---

## Simple Example

### K-Means

**Point → Find nearest centroid → Assign to cluster**

### Agglomerative

**Point → Find closest cluster → Merge clusters**

---

## Easy Memory Trick

**K-Means → Centroids 🎯**

**Agglomerative → Merge 🔗**

Both can produce **3 clusters**, but they create those clusters using different methods.

## Takeaways

- A **dendrogram** is a tree-like diagram that illustrates the arrangement and merging levels of data points in **hierarchical clustering**.

- The **SciPy library** provides a range of APIs specifically designed for **Hierarchical Clustering**.

# DBSCAN

## DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

## What is DBSCAN?

DBSCAN is an **unsupervised clustering algorithm** that groups data points based on **density**.

It identifies 3 types of points:

- 🟢 **Core Point** → Has many nearby points.
- 🟠 **Border Point** → Has fewer nearby points but is close to a core point.
- ⚪ **Outlier** → Isolated point in a low-density area.

---

## Scatter Plot Example

```text
        🟢 🟢 🟢
      🟢 🟢 🟢 🟠
    🟠 🟢 🟢 🟢
       🟢 🟢 🟠
          🟠


                         🟢 🟢 🟢
                       🟢 🟢 🟢 🟠
                     🟠 🟢 🟢 🟢
                        🟢 🟢 🟠


   ⚪


                                      ⚪

🟢 Core Point
🟠 Border Point
⚪ Outlier
```

The **dense group of points** forms a cluster.

- 🟢 Core points have enough neighbors.
- 🟠 Border points are close to core points but don't have enough neighbors themselves.
- ⚪ Outliers are isolated from the clusters.

---

## Important Parameters

### 1. `eps`

`eps` means the **maximum distance** to consider another point as a neighbor.

Example:

`eps = 0.5`

Points within this distance are considered neighbors.

### 2. `min_samples`

`min_samples` is the **minimum number of nearby points** required for a point to become a core point.

Example:

`min_samples = 5`

A point needs at least 5 points within its `eps` distance to be considered a core point.

---

## How DBSCAN Works

1. Choose `eps` and `min_samples`.
2. Find the neighbors of each point.
3. Points with enough neighbors become **core points**.
4. Points close to core points but with fewer neighbors become **border points**.
5. Isolated points become **outliers**.
6. Core points and their connected neighbors form clusters.

---

## Benefits of DBSCAN

- Good at handling **outliers**
- No need to specify the **number of clusters**
- Good at handling **weird/non-circular shapes**
- Can be faster than some other clustering methods

---

## Easy Memory Trick

**DBSCAN = Density-Based Clustering**

`eps` → **How close?**

`min_samples` → **How many points?**

🟢 Core → **Many neighbors**

🟠 Border → **Near a core point**

⚪ Outlier → **Isolated point**
