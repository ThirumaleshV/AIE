# 1.simple linear regression



## What is Linear Regression?

- **Linear Regression** helps to establish the relationship between the **dependent variable** and **independent variable(s)**.

- **Linear Regression** is a supervised machine learning algorithm used to find the relationship between input and output.

It is mainly used to **predict numerical values**.

### Example

Suppose we want to predict the **price of a house** using its **area**.

- **Area** → Independent variable (input)
- **Price** → Dependent variable (output)

The model learns the relationship between area and price and predicts the price of a new house.

---

## Simple Linear Regression

When there is **one independent variable**, we use:

$$
y = mx + b
$$

Where:

- `y` = Dependent variable / predicted value
- `x` = Independent variable / input
- `m` = Slope (or Gradient)
- `b` = Y-intercept

### House Price Example

$$
price = m \times area + b
$$

Here:

- `price` → Dependent variable
- `area` → Independent variable
- `m` → Slope
- `b` → Y-intercept

---

## Slope (m)

**Slope tells us how much `y` changes when `x` increases by 1.**

Example:

$$
price = 50 \times area + 100
$$

## What is Multiple Linear Regression?

**Multiple Linear Regression** is a supervised machine learning algorithm used to predict one numerical output using **two or more input variables**.

### Simple Example

Suppose we want to predict the **price of a house**.

The price may depend on:

- Area
- Number of bedrooms
- Age of the house

So, we use all these inputs to predict the price.

## Formula

$$
y = m_1x_1 + m_2x_2 + m_3x_3 + b
$$

Where:

- `y` = predicted output
- `x₁, x₂, x₃` = independent variables (inputs)
- `m₁, m₂, m₃` = coefficients
- `b` = intercept

### House Price Example

$$
price = m_1(area) + m_2(bedrooms) + m_3(age) + b
$$

Here:

- `price` → Dependent variable
- `area` → Independent variable
- `bedrooms` → Independent variable
- `age` → Independent variable

**Simple Linear Regression:**

Uses only **one input**.

$$
price = m(area) + b
$$

Uses **multiple inputs**.

$$
price = m_1(area) + m_2(bedrooms) + m_3(age) + b
$$

For example:

**Area + Bedrooms + Age → House Price**

## Linear Regression: Slope, Best Fit Line, Error, MAE and MSE

## 1. Slope

**Slope** tells us how much `y` changes when `x` changes.

Slope is also called **Gradient**.

$$
m = \frac{\text{Change in }y}{\text{Change in }x}
$$

### Example

Change in `y` = 3  
Change in `x` = 5

$$
m = \frac{3}{5} = 0.6
$$

So, the slope is `0.6`.

---

## 2. Equation of a Straight Line

The equation of a straight line is:

$$
y = mx + b
$$

Where:

- `y` = predicted value
- `x` = input
- `m` = slope / gradient
- `b` = intercept

### Example

$$
y = 2x + 3
$$

If `x = 4`:

$$
y = 2(4) + 3 = 11
$$

So, the predicted value is `11`.

---

## 3. Best Fit Line

In Linear Regression, we have many data points.

The model tries to find the **best line that fits these data points**.

This line is called the **Best Fit Line**.

The model finds the appropriate values for:

- `m` → slope
- `b` → intercept

The goal is to make the predicted values as close as possible to the actual values.

---

## 4. Error

**Error** is the difference between the actual value and the predicted value.

### Example

Actual value = `500`

Predicted value = `450`

$$
Error = Actual - Predicted
$$

$$
Error = 500 - 450 = 50
$$

So, the error is `50`.

---

## 5. Loss

**Loss** is a measure of how wrong the model's predictions are.

A smaller loss generally means the predictions are closer to the actual values.

MAE and MSE are two common ways to measure prediction error.

---

## 6. Mean Absolute Error (MAE)

**MAE** stands for **Mean Absolute Error**.

It calculates the **average of the absolute errors** between actual and predicted values.

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

Where:

- `y` = actual value
- `ŷ` = predicted value
- `n` = number of observations

### Example

Suppose the errors are:

    10, -20, 30

Take the absolute values:

    10, 20, 30

Then:

$$
MAE = \frac{10+20+30}{3}
$$

$$
MAE = 20
$$

So, the model's prediction is **20 units away from the actual value on average**.

### Why use absolute values?

Positive and negative errors should not cancel each other.

For example:

    10 + (-10) = 0

This would incorrectly suggest there is no error.

Instead:

    |10| + |-10| = 20

---

## 7. Mean Squared Error (MSE)

**MSE** stands for **Mean Squared Error**.

It calculates the **average of the squared errors**.

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

### Example

Suppose the errors are:

    10, -20, 30

Square each error:

    10² = 100
    (-20)² = 400
    30² = 900

Then:

$$
MSE = \frac{100+400+900}{3}
$$

$$
MSE = 466.67
$$

So, the MSE is `466.67`.

---

## 8. Why Does MSE Square the Errors?

Squaring makes **large errors much more significant**.

For example:

    Error = 2
    2² = 4

    Error = 10
    10² = 100

Therefore, MSE strongly penalizes large errors.

---

## 9. MAE vs MSE

| MAE | MSE |
|---|---|
| Uses absolute errors | Uses squared errors |
| Easy to understand | Less directly interpretable |
| Large errors are not heavily punished | Large errors are heavily punished |
| Treats errors more equally | Gives more importance to large errors |

---

## 10. Where Are MAE and MSE Used?

MAE and MSE are mainly used to **evaluate regression models**.

The process is:

    Training Data
          ↓
    Train the Model
          ↓
    Make Predictions
          ↓
    Compare Predictions with Actual Values
          ↓
    Calculate MAE / MSE
          ↓
    Evaluate Model Performance

### Example

Suppose we build a house-price prediction model.

    Actual price     = 500
    Predicted price  = 450

The error is:

    500 - 450 = 50

We do this for many houses and then calculate MAE or MSE.

These values tell us **how much error the model is making**.

---

## 11. Comparing Two Models

MAE and MSE can also be used to compare different regression models.

| Model | MAE | MSE |
|---|---:|---:|
| Model A | 20 | 500 |
| Model B | 12 | 250 |

Model B has lower MAE and MSE.

Therefore, Model B generally has **better prediction performance on the same evaluation data**.

---

We can calculate MAE and MSE using `scikit-learn`:

    mae = mean_absolute_error(y_test, y_pred)

Here:

# Derivatives and Partial Derivatives

## Derivative

A **derivative tells us how fast something is changing**.

In simple words:

> **Derivative = slope at a particular point**

Example:

$$
y = x^2
$$

Derivative:

$$
\frac{dy}{dx} = 2x
$$

---

## Partial Derivative

A **partial derivative** is used when there are **multiple variables**.

It tells us how **one variable affects the output while keeping the other variables constant**.

Example:

$$
f(x,y) = x^2 + y^2
$$

Partial derivative with respect to `x`:

$$
\frac{\partial f}{\partial x} = 2x
$$

---

## In Machine Learning

Derivatives are used in **Gradient Descent** to find the direction that reduces the **loss/error**.

### Remember

- **Derivative** → slope at a point
- **Partial Derivative** → effect of one variable while others stay constant
- **Gradient Descent** → uses derivatives to reduce loss

# Chain Rule

The **Chain Rule** is used when one function is inside another function.

In simple words:

> **Chain Rule helps us find how the final output changes by going through each step.**

### Example

$$
y = (x^2 + 1)^3
$$

Using Chain Rule:

$$
\frac{dy}{dx} = 3(x^2+1)^2 \times 2x
$$

Therefore:

$$
\frac{dy}{dx} = 6x(x^2+1)^2
$$

### Easy Way to Remember

> **Outer derivative × Inner derivative**

### In Machine Learning

The Chain Rule is important in **Gradient Descent and Backpropagation**.

It helps us find how a change in a model parameter affects the final **loss/error**.

# Gradient Descent Theory

## What is Gradient Descent?

**Gradient Descent** is an algorithm used to **reduce the loss/error of a machine learning model**.

In simple words:

> Gradient Descent finds the direction in which the loss decreases and updates the model step by step.

### Easy Example

Imagine you are standing on a hill and want to reach the lowest point.

- Hill → Loss function
- Your position → Model parameters
- Downhill direction → Gradient
- Step size → Learning rate
- Lowest point → Minimum loss

## How Gradient Descent Works

1. Start with some parameter values.
2. Calculate the loss.
3. Calculate the gradient.
4. Move in the direction that reduces the loss.
5. Repeat the process.
6. Stop when the loss becomes minimum or very small.

## Important Terms

- **Loss** → tells how wrong the model is.
- **Gradient** → tells which direction the loss increases.
- **Learning Rate** → controls how big each step is.
- **Parameters** → values such as weights and bias that the model updates.

## Learning Rate

The **learning rate** controls how big a step Gradient Descent takes.

- Too large → may jump over the minimum.
- Too small → learning becomes very slow.
- Suitable learning rate → reaches the minimum efficiently.

## In Simple Words

> **Gradient Descent = Keep changing the model parameters in the direction that reduces the error.**

## Why is it Important?

Gradient Descent is widely used in Machine Learning to **find better model parameters and minimize the loss**.

It is especially important in:

- Linear Regression
- Logistic Regression
- Neural Networks
- Deep Learning
- Backpropagation

````python
import pandas as pd
````

````python
import numpy as np
````

````python
def gradient_descent(x,y,lr=0.01,epochs=3000):
    m,b=0.0 , 0.0

    for epoch in range(epochs):
        y_pred = m*x + b
        error =y-y_pred
        cost=np.mean(error**2)

        dm=-2*np.mean(x*error)
        db=-2*np.mean(error)

        b-=db*lr
        m-=dm*lr

        print(f"m={m} , b={b} , Epoch={epochs} cost={cost}")

if __name__ == "__main__":
    x=np.array([1,2,3,4,5])
    y=np.array([5,7,9,11,13])
    gradient_descent(x,y)
   
````

````python
def gradient_descent(x, y, lr=0.1, epochs=3000):
    # Scale x and y using Min-Max Scaling
    x_min, x_max = x.min(), x.max()
    y_min, y_max = y.min(), y.max()

    x_scaled = (x - x_min) / (x_max - x_min)
    y_scaled = (y - y_min) / (y_max - y_min)

    # Initialize parameters
    b = 0.0  # Intercept
    m = 0.0  # Slope
    n = len(y_scaled)  # Number of data points

    # Perform gradient descent
    for epoch in range(epochs):
        y_pred = b + m * x_scaled  # Predicted y values
        error = y_scaled - y_pred  # Error in prediction
        cost = np.mean(error ** 2)   # Mean squared error

        # Calculate gradients
        db = -2 * np.mean(error)  # Derivative w.r.t. intercept b
        dm = -2 * np.mean(error * x_scaled)  # Derivative w.r.t. slope m

        # Update parameters
        b -= lr * db
        m -= lr * dm

        # Optional: Print cost every 100 iterations to monitor progress
        if epoch % 100 == 0:
            print(f"Epoch {epoch}: Cost = {cost}, b = {b}, m = {m}")

    # Scale back the coefficients to original scale
    b_original = b * (y_max - y_min) + y_min - m * (y_max - y_min) * x_min / (x_max - x_min)
    m_original = m * (y_max - y_min) / (x_max - x_min)

    return b_original, m_original


if __name__ == "__main__":
    df = pd.read_csv(r"..\Datasets\home_prices.csv")

    x = df["area_sqr_ft"].to_numpy()
    y = df["price_lakhs"].to_numpy()

    b, m = gradient_descent(x, y)

    print(f"Final Results: m={m}, b={b}")
````

# Why MSE (and not MAE)

## MSE and MAE

Both **MSE** and **MAE** are used to measure how wrong a regression model's predictions are.

## MAE — Mean Absolute Error

**MAE** finds the average size of the errors.

### Example

| Actual | Predicted | Error |
|---:|---:|---:|
| 100 | 90 | 10 |
| 200 | 180 | 20 |
| 300 | 330 | 30 |

$$
MAE = \frac{10+20+30}{3}
$$

$$
MAE = 20
$$

So, the model is wrong by about **20 units on average**.

---

## MSE — Mean Squared Error

**MSE** squares the errors first and then takes the average.

Using the same errors:

$$
10^2 = 100
$$

$$
20^2 = 400
$$

$$
30^2 = 900
$$

Then:

$$
MSE = \frac{100+400+900}{3}
$$

$$
MSE = 466.67
$$

### Why does MSE square the error?

Squaring makes **large errors much more important**.

Example:

$$
5^2 = 25
$$

$$
50^2 = 2500
$$

So, MSE is sensitive to **large errors and outliers**.

---

## MAE vs MSE

- **MAE** → average error
- **MSE** → average squared error
- **MAE** is easier to understand
- **MSE** gives more importance to large errors

## In Gradient Descent

MSE is commonly used with **Gradient Descent** because it is smooth and differentiable, which makes calculating gradients easier.

### Easy Way to Remember

> **MAE → How wrong is the model on average?**

> **MSE → How much should large errors be penalized?**


# RMSE — Root Mean Squared Error

**RMSE** tells us how far our predictions are from the actual values, on average.

It is similar to **MAE**, but it gives more importance to **large errors**.

## Example

Suppose:

    Actual:     100, 200, 300
    Predicted:   90, 180, 330

### Step 1: Find the Errors

    Errors: 10, 20, 30

### Step 2: Square the Errors

    10² = 100
    20² = 400
    30² = 900

### Step 3: Calculate MSE

$$
MSE = \frac{100+400+900}{3}
$$

$$
MSE = 466.67
$$

### Step 4: Calculate RMSE

$$
RMSE = \sqrt{MSE}
$$

$$
RMSE = \sqrt{466.67}
$$

$$
RMSE \approx 21.6
$$

So:

> **RMSE ≈ 21.6**

This means the model's typical prediction error is around **21.6 units**, with larger errors having more influence.

## Easy Difference

| Metric | Meaning |
|---|---|
| **MAE** | Average error |
| **MSE** | Average squared error |
| **RMSE** | Square root of MSE |

### Remember

> **MAE → Average error**

> **MSE → Average squared error**

> **RMSE → Square root of MSE**

For all three:

> **Lower value is generally better.** ✅

## R² Score

**R² (R-squared)** tells us how well a regression model explains the variation in the actual values.

## Formula

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$

Where:

- `SS_res` = error from the model's predictions
- `SS_tot` = total variation in the actual values

## SSE and SST

1. **SSE** → **Sum of Squared Errors**

2. **SST** → **Total Sum of Squares**  
   Also called **Total Sum of Squared Deviations**

## Easy Example

Actual values:

$$
10,\ 20,\ 30
$$

Predicted values:

$$
12,\ 18,\ 29
$$

### Step 1: Find the average of actual values

$$
\bar{y} = \frac{10+20+30}{3}=20
$$

### Step 2: Calculate model error

$$
SS_{res}=(10-12)^2+(20-18)^2+(30-29)^2
$$

$$
SS_{res}=4+4+1=9
$$

### Step 3: Calculate total variation

$$
SS_{tot}=(10-20)^2+(20-20)^2+(30-20)^2
$$

$$
SS_{tot}=100+0+100=200
$$

### Step 4: Calculate R²

$$
R^2=1-\frac{9}{200}
$$

$$
R^2=0.955
$$

So:

> **R² = 0.955**, meaning the model explains about **95.5% of the variation** in the actual values.

## Easy Way to Remember

> **R² compares the model's error with the total variation in the data.**

- `R² = 1` → Perfect model
- `R² = 0` → Model is no better than predicting the average
- Higher `R²` → Generally better model performance
- `R² < 0` → Model can be worse than predicting the average
- `R² = -4` → much worse than the average

# Classification Metrics

Classification metrics tell us **how well a classification model is making predictions**.

For example, in **Spam Email Detection**:

- `1` = Spam
- `0` = Not Spam

## Easy Example

Actual values:

    y_test = [1, 1, 0, 0, 1]

Predicted values:

    y_pred = [1, 0, 0, 0, 1]

Compare them:

    Actual:      1   1   0   0   1
    Prediction:  1   0   0   0   1
                 ✓   ✗   ✓   ✓   ✓

The model correctly predicted **4 out of 5**.

---

## 1. Accuracy

**Accuracy** tells us how many predictions are correct **overall**.

### Formula

$$
Accuracy = \frac{Correct\ Predictions}{Total\ Predictions}
$$

Here:

$$
Accuracy = \frac{4}{5} = 0.80
$$

So:

> **Accuracy = 80%**

### Python

    accuracy_score(y_test, y_pred)

---

## 2. Precision

**Precision** tells us:

> **When the model predicts Spam, how often is it actually Spam?**

The model predicted Spam twice:

    Predicted Spam → 1, 1

Both predictions are correct.

$$
Precision = \frac{2}{2} = 1
$$

So:

> **Precision = 100%**

### Python

    precision_score(y_test, y_pred)

---

## 3. Recall

**Recall** tells us:

> **Out of all the actual Spam emails, how many did the model find?**

Actual Spam emails:

    1, 1, 1

The model found:

    ✓, ✗, ✓

It found **2 out of 3**.

$$
Recall = \frac{2}{3} \approx 0.67
$$

So:

> **Recall ≈ 67%**

### Python

    recall_score(y_test, y_pred)

---

## 4. F1 Score

**F1 Score** gives a balance between **Precision and Recall**.

### Python

    f1_score(y_test, y_pred)

Here:

> **F1 Score ≈ 80%**

---

## Easy Way to Remember

> **Accuracy** → Overall, how many predictions are correct?

> **Precision** → When I say Spam, am I correct?

> **Recall** → Did I find all the Spam?

> **F1 Score** → Balance between Precision and Recall.

Examples:

There is no **higher** or **lower** category.

### 2. Ordinal

Categories have a **meaningful order**.

Examples:

- Size → `Small < Medium < Large`
- Rating → `Poor < Average < Good < Excellent`

> **Categorical variable = A variable that tells us "which category or group?"**

- Gender → Categorical variable
- City → Categorical variable

## Label Encoding vs One-Hot Encoding

The main question is:

> **Does the category have an order?**

Use **Label Encoding when categories have a meaningful order**.

### Example

Size:

    Small  → 0
    Medium → 1
    Large  → 2

Here:

$$
Small < Medium < Large
$$

So, the numbers represent a meaningful order.

### Use Label Encoding For

- Poor, Average, Good, Excellent
- Low, Medium, High

---

## 2. One-Hot Encoding

Use **One-Hot Encoding when categories have NO order**.

### Example

Color:

There is no relationship such as:

    Red < Blue < Green

So, using `0, 1, 2` could incorrectly make the model think there is an order.

### Use One-Hot Encoding For

- Gender
- City
- Color
- Country
- Department

---

## Easy Way to Remember

> **Ordered categories → Label Encoding**

> **Unordered categories → One-Hot Encoding**

## LabelEncoder vs get_dummies() vs OneHotEncoder

## 1. `LabelEncoder`

`LabelEncoder` converts categories into **numbers**.

Example:

    Bangalore → 0
    Chennai   → 1
    Delhi     → 2

➡️ **One column**

⚠️ It can create an artificial order:

    Bangalore < Chennai < Delhi

So, `LabelEncoder` is generally **not preferred for categorical input features**.

### Python Code

    from sklearn.preprocessing import LabelEncoder

    le = LabelEncoder()

    df["City_encoded"] = le.fit_transform(df["City"])

    print(df)

### Output

    City        City_encoded
    Bangalore   0
    Chennai     1
    Delhi       2
    Bangalore   0

---

## 2. `pd.get_dummies()`

`get_dummies()` converts categories into **multiple 0/1 columns**.

Example:

    Bangalore → 1 0 0
    Chennai   → 0 1 0
    Delhi     → 0 0 1

➡️ **Multiple columns**

### Python Code

    df_encoded = pd.get_dummies(df, columns=["City"])

    print(df_encoded)

### Output

    City_Bangalore    City_Chennai    City_Delhi
    True              False           False
    False             True            False
    False             False           True
    True              False           False

You can also use `drop_first=True`:

    df_encoded = pd.get_dummies(
        df,
        columns=["City"],
        drop_first=True
    )

    print(df_encoded)

`drop_first=True` removes the first dummy column to avoid redundant information.

---

## 3. `OneHotEncoder`

`OneHotEncoder` does the **same basic type of encoding as `get_dummies()`**.

Example:

    Bangalore → 1 0 0
    Chennai   → 0 1 0
    Delhi     → 0 0 1

But it is a **scikit-learn preprocessing tool**, so it is especially useful in **ML pipelines**.

### Python Code

    from sklearn.preprocessing import OneHotEncoder

    encoder = OneHotEncoder(sparse_output=False)

    encoded = encoder.fit_transform(df[["City"]])

    print(encoded)

### Output

    [[1. 0. 0.]
     [0. 1. 0.]
     [0. 0. 1.]
     [1. 0. 0.]]

---

## Main Difference

| Method | Converts To | Common Use |
|---|---|---|
| `LabelEncoder` | `0, 1, 2...` | Target/labels |
| `pd.get_dummies()` | Multiple `0/1` columns | Quick feature encoding |
| `OneHotEncoder` | Multiple `0/1` columns | ML preprocessing/pipelines |

---

## Easy Way to Remember

### LabelEncoder

    Bangalore → 0
    Chennai   → 1
    Delhi     → 2

➡️ **One column**

### get_dummies() / OneHotEncoder

    Bangalore → 1 0 0
    Chennai   → 0 1 0
    Delhi     → 0 0 1

➡️ **Multiple columns**

---

## Important Difference

`get_dummies()` and `OneHotEncoder` are **not exactly the same**, but they perform the same basic idea: **One-Hot Encoding**.

### `get_dummies()`

- Pandas function
- Easy and convenient
- Good for quick data preprocessing

### `OneHotEncoder`

- Scikit-learn preprocessing tool
- Useful for ML pipelines
- Can handle unknown categories using options such as `handle_unknown="ignore"`

### `LabelEncoder`

- Converts categories into numbers
- Mainly used for **target/labels**
- Generally avoid using it directly for unordered categorical input features

---

## Final Summary

> **LabelEncoder → categories become numbers in one column**

> **get_dummies() → categories become multiple 0/1 columns**

> **OneHotEncoder → categories become multiple 0/1 columns and works well with ML pipelines**

# Polynomial Regression


**Polynomial Regression** is used when the relationship between the input and output is **not a straight line**.

It adds powers of the input, such as:

$$
y = b_0 + b_1x + b_2x^2
$$

### Simple Example

Suppose we want to predict a student's score based on study hours.

A straight line may not fit the data well.

Polynomial Regression can create a curved relationship:

$$
score = b_0 + b_1(hours) + b_2(hours^2)
$$

Here:

- `hours` → input
- `score` → output
- `b₀` → intercept
- `b₁`, `b₂` → coefficients
- `x²` → polynomial feature

### Degree

The **degree** tells us the highest power of `x`.

**Degree = 2:**

$$
y = b_0 + b_1x + b_2x^2
$$

**Degree = 3:**

$$
y = b_0 + b_1x + b_2x^2 + b_3x^3
$$

### Easy Way to Remember

> **Linear Regression → straight line**

> **Polynomial Regression → curved line**

Polynomial Regression is useful when a **straight line cannot properly fit the data**.

# Overfitting and Underfitting

## Overfitting, Underfitting and Balanced Fit

## Overfitting

**Overfitting** happens when the model learns the training data **too much**, including noise.

It performs well on training data but poorly on **new/unseen data**.

> **Too complex → learns noise → poor on new data**

## Underfitting

**Underfitting** happens when the model is **too simple** and cannot learn the important patterns in the data.

It performs poorly on both training and new data.

> **Too simple → doesn't learn enough**

## Balanced Fit

A **balanced fit** happens when the model learns the important patterns from the training data and performs well on **unseen data**.

> **Good balance → learns patterns → performs well on new data**

## Easy Way to Remember

- **Overfitting** → Learns too much
- **Underfitting** → Learns too little
- **Balanced Fit** → Learns the right amount

## Simple Example: Spam Email Detection

Suppose we are building a model to predict whether an email is **Spam** or **Not Spam**.

### 1. Overfitting

The model **memorizes the training emails** instead of learning general patterns.

Example:

    Training email:
    "Win $1000 now!" → Spam

The model may memorize specific words or details from the training data.

When it receives a new email:

    "You have won a cash prize!" → Not Spam ❌

The model performs well on training data but poorly on **new emails**.

> **Overfitting → Learns too much, including noise.**

---

### 2. Underfitting

The model is **too simple** and doesn't learn enough patterns.

Example:

    "Win $1000 now!" → Spam
    "Congratulations! You won a prize!" → Not Spam ❌

The model cannot recognize different patterns of spam.

It performs poorly on both **training and new emails**.

> **Underfitting → Learns too little.**

---

### 3. Balanced Fit

The model learns the **important patterns** of spam emails without simply memorizing them.

Example:

    "Win $1000 now!" → Spam
    "Congratulations! You won a prize!" → Spam
    "Meeting is scheduled for tomorrow." → Not Spam

It can also correctly classify **new emails**.

> **Balanced Fit → Learns the right patterns and works well on new data.**


# Reasons and Remedies For Overfitting / Underfitting

## How to Handle Overfitting

Overfitting happens when the model learns the training data **too much**, including noise.

### Causes and Solutions

- **Poor model / hyperparameters** → Use a better model and tune hyperparameters.
- **Insufficient training data** → Use more training data.
- **Poor feature selection** → Select better and more useful features.
- **Inadequate validation** → Use proper validation techniques.
- **Lack of regularization** → Apply regularization.

---

## How to Handle Underfitting

Underfitting happens when the model is **too simple** and cannot learn the important patterns.

### Solutions

- **Use a more complex model** that can capture data patterns.
- **Use sufficient training data.**
- **Improve feature selection / feature engineering.**
- **Give the model sufficient training time.**
- **Use proper validation.**
- **Use appropriate regularization.**

### Easy Way to Remember

> **Overfitting → Model learns too much → Simplify / regularize / improve data**

> **Underfitting → Model learns too little → Use a better or more complex model**

# Ridge and Lasso Regression

Both **Ridge and Lasso Regression** are techniques used to **reduce overfitting** in Linear Regression.

## Ridge Regression (L2 Regularization)

Ridge Regression adds a **penalty for large coefficients**.

In simple words:

> **Ridge makes the coefficients smaller to reduce overfitting.**

### Example

Suppose Linear Regression gives:

    Area coefficient = 10
    Bedrooms coefficient = 8

Ridge may reduce them to:

    Area coefficient = 6
    Bedrooms coefficient = 5

It usually makes coefficients **smaller**, but generally does not make them exactly `0`.

### Python

    from sklearn.linear_model import Ridge

    model = Ridge(alpha=1.0)

---

## Lasso Regression (L1 Regularization)

Lasso Regression also adds a penalty, but it can make some coefficients **exactly 0**.

In simple words:

> **Lasso can remove less important features by making their coefficients 0.**

### Example

Suppose we have:

    Area coefficient = 10
    Bedrooms coefficient = 8
    Car_color coefficient = 0.02

Lasso may make:

    Area coefficient = 7
    Bedrooms coefficient = 5
    Car_color coefficient = 0

The `Car_color` feature is effectively removed from the model.

### Python

    from sklearn.linear_model import Lasso

    model = Lasso(alpha=1.0)

---

## What is `alpha = 1.0`?

`alpha` controls **how strong the regularization is**.

- Small `alpha` → weaker regularization
- Large `alpha` → stronger regularization

Example:

    alpha = 0.01 → weak regularization
    alpha = 1.0  → moderate regularization
    alpha = 10   → strong regularization

So:

> **alpha = 1.0 means the regularization strength is set to 1.0.**

The best value of `alpha` is usually chosen by testing different values and checking model performance.

---

## Ridge vs Lasso

| Ridge | Lasso |
|---|---|
| L2 Regularization | L1 Regularization |
| Makes coefficients smaller | Can make coefficients exactly 0 |
| Usually keeps all features | Can remove less important features |
| Helps reduce overfitting | Helps reduce overfitting + feature selection |

## Easy Way to Remember

> **Ridge → Shrinks coefficients**

> **Lasso → Shrinks + can remove features**

> **Alpha → Controls how strong the penalty is**

# Bias Variance Trade Off

## Bias and Variance

## 1. Bias

**Bias = model is too simple.**

Example: You have data that follows a curved pattern, but you use a straight line.

**Actual pattern:** Curved  
**Model:** Straight

The model cannot learn the pattern properly.

👉 **High Bias → Underfitting**

---

## 2. Variance

**Variance = model is too sensitive to the training data.**

The model learns the training data **too well**, including noise and small details.

Example:

**Training data → Model performs very well**  
**New data → Model performs poorly**

👉 **High Variance → Overfitting**

---

## 3. Simple Example: Spam Detection

Suppose we are building a model to detect **Spam** and **Not Spam** emails.

### High Bias → Underfitting

The model is too simple and cannot learn the important patterns.

**Example:**

Spam email → Not Spam ❌  
Not Spam email → Not Spam ✅  
Spam email → Not Spam ❌

The model cannot learn the spam patterns properly.

👉 **High Bias = Underfitting**

---

### High Variance → Overfitting

The model learns the training emails **too closely**, including noise and unnecessary details.

**Example:**

Training emails → Very accurate ✅  
New emails → Poor predictions ❌

👉 **High Variance = Overfitting**

---

## 4. Bias vs Variance

| Bias | Variance |
|------|----------|
| Model is too simple | Model is too complex |
| Misses important patterns | Learns too many details |
| High Bias → Underfitting | High Variance → Overfitting |
| Poor performance because it cannot learn enough | Good training performance but poor performance on new data |

---

## 5. Easy Way to Remember

> **High Bias → Underfitting → Model is too simple**

> **High Variance → Overfitting → Model learns too much**

---

## 6. Goal

We want to find a **balance between Bias and Variance**.

**High Bias → Underfitting**  
**High Variance → Overfitting**  
**Balanced Bias and Variance → Good Fit**

👉 **Goal = Build a model that performs well on both training and new data.**
