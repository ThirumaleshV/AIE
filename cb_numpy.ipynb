## matrix operations

```python
import numpy as np
```

```python
#  usa india china
q1=np.array([
    [200,220,250],  # Product A
    [150,180,210],  # Product B
    [300,330,360]   # Product C
])

q2=np.array([
    [209,231,259],   # Product A
    [155,192,222],   # Product B
    [310,340,375]    # Product C
])
```

```python
q1 + q2
```

```python
q2 - q1
```

```python
(q2 - q1)*100/q1  # percentage increased
```

```python
prices = np.array([
    [10, 12, 11],  # Prices for Product A in different regions
    [8,  9,  10],  # Prices for Product B
    [15, 16, 17]   # Prices for Product C
])

q1_revenue=q1 * prices
q1_revenue
```

```python
q1_discount=q1_revenue *0.2
q1_discount
```

```python
net_revenue = q1_revenue - q1_discount
net_revenue
```

```python
np.sum(q1_discount)
```

```python
### vstack, hstack
# Customer ID, Name
c = np.array([
    [101, 'Mira'],
    [102, 'Abdul'],
    [103, 'Andrea']
])
# Customer Id, Purchase Amount, Purchase Date
d = np.array([
    [101, 250.50, '2023-08-01'],
    [102, 150.00, '2023-08-02'],
    [103, 300.75, '2023-08-01']
])
```

```python
np.hstack((c,d))
```

```python
c
```

```python
e = np.array([
    [104, 'Venkat'],
    [105, 'John'],
    [106, 'Kathy'],
])
e
```

```python
np.vstack((c,e))
```
